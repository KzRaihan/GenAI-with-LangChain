### 1. Introduction of Chain

- Definition:
    - A Chain is a sequence of operations where multiple components (prompts, models, parsers, retrievers, etc.)  are connected together to perform a complete task.

    - where the output of one step automatically becomes the input of the next step.


- In simple terms,
    - Chain  is equal to the Pipeline

```python
"""          - Why Do We Need Chains?  

WITHOUT Chains (Manual Plumbing):
─────────────────────────────────
step1_output = prompt.format(topic="AI")
step2_output = llm.invoke(step1_output)
step3_output = parser.parse(step2_output)
step4_output = another_prompt.format(data=step3_output)
step5_output = llm.invoke(step4_output)

# -  5 lines of manual data passing. Fragile. Messy.


WITH Chains (Automated Pipeline):
─────────────────────────────────
chain = prompt | llm | parser | another_prompt | llm
result = chain.invoke({"topic": "AI"})

# -  One line. Clean. Composable. Reusable.

"""

# The Key Problem without Chains
# - 1. Repetitive Boilerplate
# - 2. Complex Multi-Step Workflows
# - 3. No Streaming or Batching


# The Key Benefit:
"""  
┌─────────────────────────────────────────────────────┐
│              WHY CHAINS?                            │
├──────────────────────┬──────────────────────────────┤
│ -  Composability     │ Mix and match components     │  
│-----------------------------------------------------│
│ -  Reusability       │ Define once, use everywhere  │
│-----------------------------------------------------│
│ -  Streaming         │ Real-time output built-in    │
│-----------------------------------------------------│
│ -  Batching          │ Parallel processing built-in │
│-----------------------------------------------------│
│ -  Async             │ Non-blocking execution       │
│-----------------------------------------------------│
│ -  Debugging         │ Trace every step             │
│-----------------------------------------------------│
│ -  Error Handling    │ Fallbacks and retries        │
│-----------------------------------------------------│
│ -  Monitoring        │ LangSmith integration        │
└──────────────────────┴──────────────────────────────┘

"""
```

- The Pipe Operator (|):
    - chain multiple components together into a unified pipeline.

    - eg : A | B
        - then the output of A, become the input of B


- Manual Processing, can make the code difficult to maintain.

### 3. Simple Chain Example
- Idea:
    - step1: Give a prompt form user

    - step2: process the user prompt by using LLMs

    - Step3: return the llm response based on user prompt

- Create a pipeline or chain that execute the above steps into one single line of code.

```python
# import necessary libraries
from dotenv import load_dotenv
from langchain_openai import ChatOpenAI
from langchain_core.prompts import PromptTemplate
from langchain_core.output_parsers import StrOutputParser

# load the api key form .env file
load_dotenv()

# Define model
model = ChatOpenAI()

# create prompt
prompt = PromptTemplate(
    template = "Explain {topic} in one sentence.",
    input_variables=["topic"]
)

# define string output parser -> to organize the output
parser = StrOutputParser()


# Now Define the Chain ( LangChain Expression Language (LCEL))
chain = prompt | model | parser

response = chain.invoke(
    {
        "topic": "quantum computing"
    }
)

print(response)

# for visualization the chain
chain.get_graph().print_ascii()

```

```python
"""   What Happens internally? 

chain = prompt | llm | parser

Step 1: prompt.invoke({"topic": "quantum computing"})
   → [HumanMessage("Explain quantum computing in one sentence.")]

Step 2: llm.invoke([HumanMessage("Explain quantum computing...")])
   → AIMessage(content="Quantum computing uses qubits...")

Step 3: parser.invoke(AIMessage(content="Quantum computing uses qubits..."))
   → "Quantum computing uses qubits..."  (clean string)

Final Output: "Quantum computing uses qubits..."

"""
```

```python
from langchain_openai import ChatOpenAI
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser


# ---------------------------------------------------------
# 1. Create the model
# ---------------------------------------------------------

model = ChatOpenAI(
    model="gpt-4o"
)


# ---------------------------------------------------------
# 2. Create the prompt
# ---------------------------------------------------------

prompt = ChatPromptTemplate.from_template(
    """
    You are an AI instructor.

    Explain {topic} to a beginner
    using a simple example.
    """
)


# ---------------------------------------------------------
# 3. Create the output parser
# ---------------------------------------------------------

parser = StrOutputParser()


# ---------------------------------------------------------
# 4. Connect the components
# ---------------------------------------------------------

chain = prompt | model | parser


# ---------------------------------------------------------
# 5. Execute the complete chain
# ---------------------------------------------------------

result = chain.invoke({
    "topic": "LangChain"
})


# ---------------------------------------------------------
# 6. Display final result
# ---------------------------------------------------------

print(result)
```

### KeyTakeaway
- A Chain in LangChain is a composable workflow that connects multiple components—such as prompts, models, retrievers, tools, and output parsers—so that data flows automatically from one step to the next.
