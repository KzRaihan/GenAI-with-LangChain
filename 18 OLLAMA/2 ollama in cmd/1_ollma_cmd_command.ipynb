### 1. Phase 3 — Ollama CLI

- All ollama command of (CLI):

    - 1. Check the ollama version:
        - ollama --version

    - 2. Download model From ollama repo:
        - ollama pull model id
        - eg: ollama run mistral-small3.1
        
        - go the model section of ollma
            - link: https://ollama.com/search

    - 3. show the all model list
        - ollama list or ollama ls
        
        - eg:  name         ID         Size       Modified 
            - qween3:8b   15454e       5.2 GB       19 hours ago     
            - gemma3:4b
            - lamma3.2:1b

    - 4. how to run model (fetch the model from storage and set in the working memory(RAM, VRAM))
        - ollama run model name
        - eg: ollama run qween3:8b

        - >> here, we ask any question?

    - 5. Exit the current model:
        - /bye
    
    - 6. show the available command for specific model 
        - ollama run qween3:8b

        - >>> /show

        - >>> /show info


        - >>> /show parameters

        - set the parameters (show the list of editable parameters)
            - >>> /set
            - >>> /set parameters
            - >>> /set parameter top_p 0.99
            - >>> /show parameters


- for other command:
    - link: https://docs.ollama.com/cli

- Key Rule of Thump:
    - command line command are use, most of time for experimental purpose.
