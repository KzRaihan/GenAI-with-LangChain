## scratch implementation __ without chian and with chai

### without chain__prediction

```python
# LLM response template class
import random

class NakliLLM:

  def __init__(self):
    print('LLM created')

  def predict(self, prompt):

    response_list = [
        'Dhaka is the capital of Bangladesh',
        'BPL is a cricket league',
        'AI stands for Artificial Intelligence'
    ]

    return {'response': random.choice(response_list)}
```

```python
# template or prompt class
class NakliPromptTemplate:

  def __init__(self, template, input_variables):
    self.template = template
    self.input_variables = input_variables

  def format(self, input_dict):
    return self.template.format(**input_dict)
```

```python
# object of template class
template = NakliPromptTemplate(
    template='Write a {length} poem about {topic}',
    input_variables=['length', 'topic']
)
```

```python
# formatting the prompt by using 
prompt = template.format({'length':'short','topic':'Bangladesh'})
```

```python
# initialize the llm
llm = NakliLLM()
```

```python
# predict
llm.predict(prompt)
```

### With chain__prediction

```python
# chain class
class NakliLLMChain:

  def __init__(self, llm, prompt):
    self.llm = llm
    self.prompt = prompt

  def run(self, input_dict):

    final_prompt = self.prompt.format(input_dict)
    result = self.llm.predict(final_prompt)

    return result['response']
```

```python
template = NakliPromptTemplate(
    template='Write a {length} poem about {topic}',
    input_variables=['length', 'topic']
)
```

```python
llm = NakliLLM()
```

```python
chain = NakliLLMChain(llm, template)
```

```python
# prediction using chain
chain.run({'length':'short', 'topic': 'Bangladesh'})
```
