### 1. Introduction of Runnable 

- Definition:
    - A Runnable is a LangChain component that can receive an input, perform an operation, and produce an output in a standardized way.


    - A Runnable is an object that follows LangChain's standard execution interface.


```python
""" - Runnable Mental Model
        ┌──────────────────────────┐
        │        Runnable          │
        │                          │
        │   Input → Processing     │
        │              ↓           │
        │           Output         │
        └──────────────────────────┘

- LangChain components follow the same interface, they can be connected together.

"""
```

```python
""" 
┌─────────────────────────────────────────────────────────────┐
│              THE RUNNABLE PHILOSOPHY                        │
│                                                             │
│  In LangChain, EVERYTHING is a Runnable:                    │
│                                                             │
│  -  ChatPromptTemplate   → is a Runnable                    │
│  -  ChatOpenAI           → is a Runnable                    │
│  -  StrOutputParser      → is a Runnable                    │
│  -  VectorStoreRetriever → is a Runnable                    │
│  -  A Python function    → can BECOME a Runnable            │
│  -  A dictionary         → can BECOME a Runnable            │
│  -  A chain of chains    → is a Runnable                    │
│                                                             │
│  Because they ALL implement the SAME interface,             │
│  they ALL work together seamlessly with the | operator.     │
│                                                             │
│  This is like USB — every device uses the same port,        │
│  so everything plugs into everything.                       │
└─────────────────────────────────────────────────────────────┘

"""
```

### 2. Why Do We Need Runnables?

- in manually manage:
    - input formats
    - output formats
    - function calls
    - data transformations
    - execution
    - streaming
    - batching

```python
"""      - LangChain solves this problem with a common abstraction:

                    Runnable
                       │
          ┌────────────┼────────────┐
          ↓            ↓            ↓
       Prompt        Model        Parser


- These components can communicate through a standardized interface.
"""
```

```python
"""         - Without VS With Runnable 


# Without the Runnable interface, you'd need custom glue code:
    messages = prompt.format(topic="AI")          # Custom method
    response = llm.generate(messages)             # Different method!
    text = parser.extract(response.generations)   # Yet another method!

# With the Runnable interface, EVERYTHING uses the same methods:
    chain = prompt | llm | parser                 # All Runnables!
    result = chain.invoke({"topic": "AI"})        # One method!

"""
```

### 3. Taxonomy of All Runnable Types

```python
"""     - 3. Taxonomy of All Runnable Types

┌──────────────────────────────────────────────────────────────────┐
│                   COMPLETE RUNNABLE TAXONOMY                      │
│                                                                  │
│  ┌────────────────────────────────────────────────────────────┐  │
│  │  COMPOSITION RUNNABLES (Combine other Runnables)           │  │
│  │  ├── RunnableSequence      → A | B | C (pipe chain)       │  │
│  │  ├── RunnableParallel      → {a: A, b: B} (fan-out)       │  │
│  │  └── RunnableBranch        → Conditional routing (if/else) │  │
│  └────────────────────────────────────────────────────────────┘  │
│                                                                  │
│  ┌────────────────────────────────────────────────────────────┐  │
│  │  TRANSFORMATION RUNNABLES (Modify data flow)               │  │
│  │  ├── RunnableLambda        → Wrap any Python function      │  │
│  │  ├── RunnablePassthrough   → Pass input unchanged          │  │
│  │  ├── RunnableEach          → Apply chain to each item      │  │
│  │  └── RunnableMap           → Map inputs to keys            │  │
│  └────────────────────────────────────────────────────────────┘  │
│                                                                  │
│  ┌────────────────────────────────────────────────────────────┐  │
│  │  RESILIENCE RUNNABLES (Error handling)                     │  │
│  │  ├── RunnableWithFallbacks → Try backup on failure         │  │
│  │  └── RunnableRetry         → Auto-retry N times            │  │
│  └────────────────────────────────────────────────────────────┘  │
│                                                                  │
│  ┌────────────────────────────────────────────────────────────┐  │
│  │  CONFIGURATION RUNNABLES (Modify behavior)                 │  │
│  │  ├── RunnableBinding       → Bind config/kwargs            │  │
│  │  └── RunnableConfigurable  → Runtime configuration         │  │
│  └────────────────────────────────────────────────────────────┘  │
│                                                                  │
│  ┌────────────────────────────────────────────────────────────┐  │
│  │  COMPONENT RUNNABLES (LangChain building blocks)           │  │
│  │  ├── ChatPromptTemplate    → Prompt formatting             │  │
│  │  ├── ChatOpenAI            → LLM inference                 │  │
│  │  ├── StrOutputParser       → Output parsing                │  │
│  │  ├── VectorStoreRetriever  → Document retrieval            │  │
│  │  └── ... (100+ more)                                       │  │
│  └────────────────────────────────────────────────────────────┘  │
└──────────────────────────────────────────────────────────────────┘

"""
```

### 4. Example to understand why and how Runnable work?
- For Runnable purpose, the chian are work.
    - how to Chain work, we must know the how to runnable work.

#### Why Runnable Exit?

- History of LangChain:
    - 1. LangChain is framework, that easily interact different type of LLMs(openai, google, anthropic)

    - 2. Apart of only interact the LLm, there are so many component are work to create an application.

```python
"""   - For Example : To Create a "PDF Reader" 
- To Create this application, we perform different types of tasks

- workflow:
    - [ PDF_load] -> [split] --> [embedding] --> [vectorDB] --> [Retrieval] --> [LLM]--> [parser] --> [result].


- For here, To create a PDF Reader application, there are different type of components that we want to interact.

- Therefore, the langchain team, build different types of component, so that developer can easily can implement and develop there application.

- They create various components such as,
    
    - 1. for Load Document(PDF, docs, file) -> document_loader component

    - 2. for split Document -> text_split component

    - 3. for embedding Generate -> embedding model 

    - and so on...  

- Developer use these components to interact each other and easy build application. 

- Runnable is help to do this more easy to implement. 

"""
```

```python
### Example without Runnable -> pdf reader app
# this is the old version of langchain implementation

from langchain.llms import OpenAI
from langchain.embeddings import OpenAIEmbeddings

from langchain.document_loaders import TextLoader
from langchain.text_splitter import RecursiveCharacterTextSplitter

# for Vector DataBase
from langchain.vectorstores import FAISS


# 1. Load the document
loader = TextLoader("docs.txt") # paht of docs
documents = loader.load()

# 2. Split the text into smaller chunks
text_splitter = RecursiveCharacterTextSplitter(chunk_size = 500, chunk_overlap = 50)
docs = text_splitter.split_documents(documents)

# 3. Convert text to embeddings and store in FAISS vector database
vectorstore = FAISS.from_documents(docs,OpenAIEmbeddings())

# 4. Create a retriever (fetches relevant documents)
retriever = vectorstore.as_retriever()

# 5. Manually Retrieve Relevant Documents
query = "What are the key takeaways from the document?"
retriever_docs = retriever.get_relevant_documents(query)

# 6. Combine Retrieved Text into a Single Prompt
retrieved_text = "\n".join([doc.page_content for doc in retriever_docs])

# 7. Initialize the LLm
llm = OpenAI(
    model_name = "gpt-3.5-turbo",
    temperature = 0.7
)

# 8. Manually Pass Retriever Text to llm
prompt = f"Based on the following text, answer the question: {query}\n\n {retrieved_text}"

answer = llm.predict(prompt)

# 9.print the answer
print("Answer: ",answer)
```

```python
# with chain -> pdf reader app

# this is the old version of langchain implementation

from langchain.llms import OpenAI
from langchain.embeddings import OpenAIEmbeddings

from langchain.document_loaders import TextLoader
from langchain.text_splitter import RecursiveCharacterTextSplitter

# for Vector DataBase
from langchain.vectorstores import FAISS

# for chain
from langchain.chain import RetrievalQA


# 1. Load the document
loader = TextLoader("docs.txt") # paht of docs
documents = loader.load()

# 2. Split the text into smaller chunks
text_splitter = RecursiveCharacterTextSplitter(chunk_size = 500, chunk_overlap = 50)
docs = text_splitter.split_documents(documents)

# 3. Convert text to embeddings and store in FAISS vector database
vectorstore = FAISS.from_documents(docs,OpenAIEmbeddings())

# 4. Create a retriever (fetches relevant documents)
retriever = vectorstore.as_retriever()


# 5. Initialize the LLm
llm = OpenAI(
    model_name = "gpt-3.5-turbo",
    temperature = 0.7
)

# 6. Create an chain
qa_chain = RetrievalQA.from_chain_type(llm=llm, retriever=retriever)

# 7. Ask a question
query = "what is the key takeaways from the document"

answer = qa_chain.run(query)

# 8.print the answer
print("Answer: ",answer)

```
