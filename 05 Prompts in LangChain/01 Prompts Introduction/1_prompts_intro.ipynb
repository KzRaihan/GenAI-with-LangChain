### 1. Introduction of Prompts Component in LangChain

- Definition of Prompts:
    - Prompts are the input instructions or queries given to a model to guide it's output.


- Types of Prompts:
    - 1. Text-base Prompt
        - Textual prompt (only text instruction)
    
    - 2. Multi-modal Prompts
        - Text + image
        - Audio + Text
        - Video + text


- Why are Prompts Important?
    - Prompts are important because they guide the AI model to understand what we want and how we want the response.

    - Improve accuracy: Well-designed prompts can produce more relevant and useful responses.

    - A good prompt helps the model understand what to do, how to do it, and what kind of output we expect.

### 3. Static vs Dynamic Prompts

- Static Prompts:
    - A static prompt is a fixed prompt where the text remains the same every time we send it to the model.

    - Static prompt = the instruction stays the same.



```python
# Example of Static Prompt
from langchain_openai import ChatOpenAI

# Create the model
model = ChatOpenAI(
    model = "gpt-4o"
)

# Static prompt
prompt = """
Explain Machine Learning to a beginner.
Give one real-world example.
"""

# Send the prompt to the model
response = model.invoke(prompt)

# Display the response
print(response.content)
```

#### Common use cases
# ------------------------------------------------------------------
- Simple question-answering

- Testing an LLM

- One-time experiments

- Fixed system instructions

- Simple prototypes

- Tasks where the input does not change


### What is the Key Drawback of Static Prompts?
# ------------------------------------------------------------------
- Main Problem:
    - A static prompt is not flexible when the input changes.

```python
# Problem of Static Prompt
"""
prompt1 = "Explain Machine Learning to a beginner."

prompt2 = "Explain Deep Learning to a beginner."

prompt3 = "Explain Computer Vision to a beginner."  

"""

# NOTE -> This leads
    # Repeated code
    # Less flexibility
    # Difficult maintenance
    # Poor scalability
    # Same prompt structure has to be rewritten
```

### Dynamic Prompts — The Solution
# ------------------------------------------------------------------
- Definition:
    - A dynamic prompt is a reusable prompt that contains variables/placeholders whose values can change depending on the user's input or application requirements.


- Example:
    - "Explain {topic} to a beginner."

    - where,
        - {topic} = Machine Learning
        - {topic} = Deep Learning
        - {topic} = Computer Vision
        - {topic} = NLP



- PromptTemplate is a LangChain class used to create reusable and dynamic prompts.

```python
# Dynamic Prompt Approach in LangChain
from langchain_core.prompts import PromptTemplate
from langchain_openai import ChatOpenAI

# Create the chat model
model = ChatOpenAI(model="gpt-4o")

# Create a reusable dynamic prompt template
prompt = PromptTemplate.from_template(
    """
    Explain {topic} to a beginner.

    Include:
    1. A simple definition
    2. One real-world example
    3. Why it is important
    """
)

# Provide a value for the {topic} variable
formatted_prompt = prompt.invoke({
    "topic": "Machine Learning"
})

# Send the formatted prompt to the model
response = model.invoke(formatted_prompt)

# Display the response
print(response.content)
```

### Key takeaway:

- Prompt = instructions + context + task + constraints that guide the model toward the desired output.

- Static Prompt: Fixed instruction → simple but less flexible.

- Dynamic Prompt: Reusable template + variables → flexible and scalable.
