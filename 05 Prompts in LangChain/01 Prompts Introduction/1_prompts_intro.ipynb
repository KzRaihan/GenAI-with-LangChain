### 1. Introduction of Prompts Component in LangChain

- Definition of Prompts:
    - Prompts are the input instructions or queries given to a model to guide it's output.


- Types of Prompts:
    - 1. Text-base Prompt
        - Textual prompt (only text instruction)
    
    - 2. Multi-modal Prompts
        - Text + image
        - Audio + Text
        - Video + text


- Why are Prompts Important?
    - Prompts are important because they guide the AI model to understand what we want and how we want the response.

    - Improve accuracy: Well-designed prompts can produce more relevant and useful responses.

    - A good prompt helps the model understand what to do, how to do it, and what kind of output we expect.



### 3. Static vs Dynamic Prompts

- Static Prompts:
    - A static prompt is a fixed prompt where the text remains the same every time we send it to the model.

    - Static prompt = the instruction stays the same.



```python
# Example of Static Prompt
from langchain_openai import ChatOpenAI

# Create the model
model = ChatOpenAI(
    model = "gpt-4o"
)

# Static prompt
prompt = """
Explain Machine Learning to a beginner.
Give one real-world example.
"""

# Send the prompt to the model
response = model.invoke(prompt)

# Display the response
print(response.content)
```

#### Common use cases
# ------------------------------------------------------------------

- Testing an LLM

- Fixed system instructions

- Tasks where the input does not change


### What is the Key Drawback of Static Prompts?
# ------------------------------------------------------------------
- Main Problem:
    - A static prompt is not flexible when the input changes.

```python
# Problem of Static Prompt
"""
prompt1 = "Explain Machine Learning to a beginner."

prompt2 = "Explain Deep Learning to a beginner."

prompt3 = "Explain Computer Vision to a beginner."  

"""

# NOTE -> This leads
    # Repeated code
    # Less flexibility
    # Difficult maintenance
    # Poor scalability
    # Same prompt structure has to be rewritten
```

### Dynamic Prompts — The Solution
# ------------------------------------------------------------------
- Definition:
    - A dynamic prompt is a reusable prompt that contains variables/placeholders whose values can change depending on the user's input or application requirements.


- Example:
    - "Explain {topic} to a beginner."

    - where,
        - {topic} = Machine Learning
        - {topic} = Deep Learning
        - {topic} = Computer Vision
        - {topic} = NLP



- PromptTemplate is a LangChain class used to create reusable and dynamic prompts.

```python
# Dynamic Prompt Approach in LangChain
from langchain_core.prompts import PromptTemplate
from langchain_openai import ChatOpenAI

# Create the chat model
model = ChatOpenAI(model="gpt-4o")

# Create a reusable dynamic prompt template
prompt = PromptTemplate.from_template(
    """
    Explain {topic} to a beginner.

    Include:
    1. A simple definition
    2. One real-world example
    3. Why it is important
    """
)

# Provide a value for the {topic} variable
formatted_prompt = prompt.invoke({
    "topic": "Machine Learning"
})

# Send the formatted prompt to the model
response = model.invoke(formatted_prompt)

# Display the response
print(response.content)
```

### 3. Why LangChain Prompt Template?

- Prompt Template
    - A PromptTemplate in LangChain is a structured way to to create prompts dynamically by inserting Variables into a predefine template.

    - Instead of hardcoding prompts, PromptTemplate allows us to define placeholders that can be filled in at filled in a runtime with different inputs.

    - This makes it reusable, flexible, and easy to manage, especially when working with dynamic user inputs or automated workflows.


- Why use PromptTemplate over f-string?
    - 1. Default Validation.

    - 2. Reusable

    - 3. LangChain Ecosystem.



```python
# 1. Default Validation.
# ----------------------------------------
#  -> Dynamic Prompt Example with streamlit
# ----------------------------------------
from langchain_openai import ChatOpenAI
from dotenv import load_dotenv

# streamlit for user interface
import streamlit as st

# for Dynamic Prompt Templates 
from langchain_core.prompts import PromptTemplate



# load the openai api from .env files
load_dotenv()

# Create the model
model = ChatOpenAI(
    model = "gpt-4o"
)

# Define the streamlit app header name
st.header('Reasearch Tool')



# Define the user interface for prompts(streamlit)
paper_input = st.selectbox( "Select Research Paper Name", ["Attention Is All You Need", "BERT: Pre-training of Deep Bidirectional Transformers", "GPT-3: Language Models are Few-Shot Learners", "Diffusion Models Beat GANs on Image Synthesis"] )

style_input = st.selectbox( "Select Explanation Style", ["Beginner-Friendly", "Technical", "Code-Oriented", "Mathematical"] ) 

length_input = st.selectbox( "Select Explanation Length", ["Short (1-2 paragraphs)", "Medium (3-5 paragraphs)", "Long (detailed explanation)"] )


# template for Dynamic Prompts
template = PromptTemplate(
    template = """ 
Please summarize the research paper titled "{paper_input"} with the following specifications:

Explanation Style: {style_input}  
Explanation Length: {length_input}  

1. Mathematical Details:     
    - Include relevant mathematical equations if present in the paper.     
    - Explain the mathematical concepts using simple, intuitive code snippets where applicable.  

2. Analogies:     
    - Use relatable analogies to simplify complex ideas.  
    
If certain information is not available in the paper, respond with: "Insufficient information available" instead of guessing.  

Ensure the summary is clear, accurate, and aligned with the provided style and length.

""",
input_variables = ['paper_input', 'style_input', 'length_input'],
validation_template = True # automatically check whether all input variable are initialize or not? if not then arise error message.

)



# file the placeholders
prompt = template.invoke({
    'paper_input': paper_input,
    'style_input': style_input,
    'length_input': length_input
}

)

if st.button:
    response = model.invoke(prompt)
    st.write(response.content)

```

#### 3. LangChain Ecosystem.
- using chain component

```python
from langchain_openai import ChatOpenAI
from dotenv import load_dotenv
import streamlit as st
from langchain_core.prompts import PromptTemplate,load_prompt

load_dotenv()
model = ChatOpenAI()

st.header('Reasearch Tool')

paper_input = st.selectbox( "Select Research Paper Name", ["Attention Is All You Need", "BERT: Pre-training of Deep Bidirectional Transformers", "GPT-3: Language Models are Few-Shot Learners", "Diffusion Models Beat GANs on Image Synthesis"] )

style_input = st.selectbox( "Select Explanation Style", ["Beginner-Friendly", "Technical", "Code-Oriented", "Mathematical"] ) 

length_input = st.selectbox( "Select Explanation Length", ["Short (1-2 paragraphs)", "Medium (3-5 paragraphs)", "Long (detailed explanation)"] )

template = load_prompt('template.json')



if st.button('Summarize'):
    chain = template | model
    result = chain.invoke({
        'paper_input':paper_input,
        'style_input':style_input,
        'length_input':length_input
    })
    st.write(result.content)
```

### Key takeaway:

- Prompt = instructions + context + task + constraints that guide the model toward the desired output.

- Static Prompt: Fixed instruction → simple but less flexible.

- Dynamic Prompt: Reusable template + variables → flexible and scalable.
