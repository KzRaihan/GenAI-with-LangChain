### 1. Models
- Models:
    - Models are the core interfaces through which we interact with AI model.
    - The reasoning engine and core "brain" of any LangChain application.
    - Example: GPT, Gemini, Claude, Llama.


- Types of Models
    - 1. Language Model or LLMs
        - input: Text
        - output: Text

    - 2. Embedding Model
        - input: Text
        - output: vector
        - use case: Semantic Search.


- Short note: 
    - LangChain model Component is a interface, that communicate different types of AI models into Standard Format.


```python
""" 
User Input
    ↓
different types of AI Models
    ↓
AI Response

"""
```

### 2. Prompts
- Prompts are instructions or inputs given to the model.

- Example: 
    - Explain {topic} in simple language.
    - Explain "Machine Learning" in simple language.


- Types of Prompt

    - 1. Dynamic and Reusable Prompts

        - A dynamic prompt uses variables/placeholders, so the same prompt can be reused with different inputs.
        
        - Example:
            - PromptTemplate.from_template("Summarize {topic} in {emotion} tone")

            - prompt.format(topic="Python", emotion="fun")

        - Short note: A reusable prompt template that changes based on input variables.

    
    - 2. Role-Based Prompts
        - A role-based prompt assigns a specific role or persona to the AI.

        - Example:
            - You are an experienced Python instructor.
            - Explain Python decorators to a beginner.

        
        - Short note: Gives the AI a specific role or expertise to influence its response.



    - 3. Few-Shot Prompts
        - A few-shot prompt provides a few examples of the desired input-output behavior before asking the model to solve a new task.

        - Example:
            - Input: I love this movie.
            - Output: Positive

            - Input: This product is terrible.
            - Output: Negative

            - Input: The service was excellent.
            - Output: (Positive)

        
        - Short note: Provides a few examples to teach the model the expected pattern.



- Short note: 
    - Prompts tell the model what to do and how to respond.


### 3. Chains

- A chain connects multiple operations together in a defined sequence.

-  Pre-defined, sequential pipelines that link multiple components together.
    - (e.g., Prompt ➔ Model ➔ Output Parser)

- Types of Chains:
    - 1. Parallel Chain

    - 2. Conditional Chain

    - Using Chain we create complex Architecture.

- Short note
    - Chain = A sequence of connected operations that work together to complete a task.



```python
""" 
- Chain Example: Translation and Summarization the input text.

    Prompt(Text)
        ↓
    (LLM) Language Model (use for translation)
        ↓
    Output Parser
        ↓
    (LLM) Language Model (use for summarization)
        ↓
    Final Output


- We can't define a pipeline for all operations then we want to do in manually , but chain is a way to define a pipeline for specific operations.
- One component output is automatically passed to the next component as input.

"""
```

#### 1. Parallel Chain

- Parallel Chain: 
    - A Parallel Chain runs multiple tasks at the same time using the same or related input.

    - It is useful when tasks are independent of each other.

    - This can make an application faster than executing each task one after another.

```python
"""   
                User Input
                    ↓
          ┌─────────┴─────────┐
          ↓                   ↓
    Generate Summary      Generate Keywords
          ↓                   ↓
          └─────────┬─────────┘
                    ↓
              Final Result


         

Short note: A Parallel Chain executes multiple independent tasks simultaneously.
"""
```

#### 2. Conditional Chain
- Conditional Chain:
    - A Conditional Chain chooses the next step based on a condition or result.
    
    - It is useful when different inputs require different processing paths.

```python
"""  

              User Question
                    ↓
              Check Question
                    ↓
          ┌─────────┴─────────┐
          ↓                   ↓
       Math?                General?
          ↓                   ↓
 Calculator Chain       LLM Chain
          ↓                   ↓
          └─────────┬─────────┘
                    ↓
                Answer


                
- Short note: A Conditional Chain selects a processing path based on a condition.
"""
```

### 4. Memory

- Memory:
    - Memory allows an application to retain relevant information from previous interactions or workflow steps.
    
    - It is useful for conversational applications and stateful workflows.


- Common Strategies:
    - 1. Buffer Memory: 
        - Stores the entire raw conversation history.

    - 2. Window Memory: 
        - Keeps only the last K conversational turns to save tokens.

    - 3. Summary Memory: 
        - Uses an LLM in the background to continuously compress and summarize old conversations.

    - 4. Custom Memory:
        - For advanced use cases, we can store specialized state in a custom memory class.
            - (eg. the user's preferences or key facts about them)

- Key Takeaway: 
    - Overcomes the "stateless" nature of LLM APIs by automatically injecting past conversation history into subsequent model calls.

```python
# Example of Memory in LangChain
""" 

User: My name is Rahim.
       ↓
     Memory
       ↓
User: What is my name?
       ↓
AI: Your name is Rahim.

"""
```

### 5. Indexes
- Indexes:
    - Indexes Connect our application to external Knowledge, such as PDFs, websites or database.

    - They are commonly used in RAG (Retrieval-Augmented Generation) applications.
    - They involve processes such as document loading, splitting, embedding, and storing.


- Sub-Components:
    - 1. Document Loaders: 
        - Ingest raw data from 100+ sources (PDFs, CSVs, URLs, SQL, Notion).

    - 2. Text Splitters: 
        - Chunk long documents into semantically coherent segments within token limits.

    - 3. Vector Stores: 
        - Store chunk embeddings for high-speed similarity search (e.g., Chroma, FAISS, Pinecone).

    - 4. Retrievers: 
        - Query engines that return the most relevant document chunks based on a user's prompt.

- Short note: 
    - Indexes organize external information so that relevant knowledge can be efficiently retrieved for the model.

    -  The bridge that connects LLMs to our private, real-time, domain-specific data.


### 6. AI Agents

- AI Agents
    - Autonomous decision-makers that use an LLM not just to generate text, but to reason about which actions to take and in what order.

    - User Query⟶Reason/Think⟶Select Tool⟶Observe Result⟶Final Answer

- Uses Tools: 
    - Connected to external APIs, calculators, search engines (Tavily/Google), databases, or Python code interpreters.


- Key Takeaway: 
    - Unlike fixed Chains, Agents dynamically determine their own execution path at runtime based on the problem at hand.

```python
""" 

User Question
      ↓
    Agent
      ↓
Choose Tool
      ↓
Execute Tool
      ↓
Get Result
      ↓
Final Answer

"""
```
