### 1. Introduction of Tools

- Definition:
    - A tool is just a Python function (or API) that is packaged in a way the LLM can understand and call when needed.


- In LangChain:
    - A Tool in LangChain is an interface that connects an LLM to external computational capabilities—such as databases, APIs, code interpreters, search engines, or custom Python functions.



### 2. Why do LLMs need Tools? (The Limitations of Pure LLMs)

- LLMs (like GPT) are great are:
    - 1. Reasoning
        - think 
        - like brain

    - 2. Language generation
        - speech


- The Limitations of Pure LLMs:
    - They can't do things like
        - Access live data(weather, news)
        - Do reliable math
        - Call APIs
        - Run code
        - Interact with a database.



- Key Rule of Thump:
    - An LLM without tools = A person who can only talk and speech but no hands and legs.
    
    - An LLM with tools = A person who can also make phone calls, 
                    use a calculator, open files, etc. (they have legs and hands)

```python
"""
┌──────────────────────────────────────────────────────────────────┐
│                   WHY LLMs REQUIRE TOOLS                         │
│                                                                  │
│  PROBLEM                  WITHOUT TOOLS         WITH TOOLS       │
│  ──────────────────────────────────────────────────────────────  │
│  Current Data             Knowledge cutoff      Live API calls   │
│  ──────────────────────────────────────────────────────────────  │
│                                                                  │
│  Math & Arithmetic        Hallucinates figures  Calculator/NumPy │
│  ──────────────────────────────────────────────────────────────  │
│                                                                  │
│  Deterministic Logic      Inconsistent code     Python REPL      │     
│  ──────────────────────────────────────────────────────────────  │
│                                                                  │
│  Stateful Actions         Cannot act            Send emails/SQL  │ 
│  ──────────────────────────────────────────────────────────────  │
│                                                                  │
│  Private/Siloed Data      No access             Retriever Tools  │
└──────────────────────────────────────────────────────────────────┘

"""



""" 
Without Tools:
LLM Input → LLM Brain → Text Output
(No ability to call APIs, run code, access databases, etc.)

With Tools:
LLM Input → LLM Brain → Decides to use Tool
                             ↓
                      [Tool Executes]
                             ↓
                      Result back to LLM
                             ↓
                      LLM Generates Final Answer

"""
```

### 3. Complete Taxonomy of Tools in LangChain

- Tools are two type 
    - 1. Built-in tools

    - 2. Custom Tools


- 1. Built-in tools
    - A built-in tool is a tool that LangChain already provides for us —it’s pre-built, production ready, and requires minimal or no setup.

    - We don't have to write the function logic ourself, just import and use it.

    - Some Build in tools
        - Web Search Tools → Tavily, Google Search, DuckDuckGo
        - API Connectors → RequestsTool, GraphQL, OpenAPI
        - Wikipedia summary -> WikipediaQueryRun
        - Run raw python code -> PythonREPLTool
        - Run Shell commends -> ShellTool
        - Send email vs Gmail -> GmailSendMessageTools
        - and so on..
          
    

```python
""" 
┌──────────────────────────────────────────────────────────────────┐
│                   TOOL TAXONOMY IN LANGCHAIN                     │
│                                                                  │
│  ┌────────────────────────────────────────────────────────────┐  │
│  │  CUSTOM TOOLS                                              │  │
│  │  ├── @tool Decorator             → Easiest, most common    │  │
│  │  ├── StructuredTool.from_function→ Explicit Pydantic schema│  │
│  │  └── BaseTool Subclass           → Full lifecycle control  │  │
│  └────────────────────────────────────────────────────────────┘  │
│                                                                  │
│  ┌────────────────────────────────────────────────────────────┐  │
│  │  RETRIEVER AS A TOOL                                       │  │
│  │  └── create_retriever_tool       → Turns any RAG retriever │  │
│  │                                    into an on-demand tool  │  │
│  └────────────────────────────────────────────────────────────┘  │
│                                                                  │
│  ┌────────────────────────────────────────────────────────────┐  │
│  │  COMMUNITY / THIRD-PARTY TOOLS                             │  │
│  │  ├── Search Tools       → Tavily, Google Search, DuckDuckGo│  │
│  │  ├── Computation Tools  → PythonREPL, WolframAlpha         │  │
│  │  ├── API Connectors     → RequestsTool, GraphQL, OpenAPI   │  │
│  │  └── File & OS Tools    → ReadFileTool, DirectoryListTool  │  │
│  └────────────────────────────────────────────────────────────┘  │
│                                                                  │
│  ┌────────────────────────────────────────────────────────────┐  │
│  │  TOOLKITS (Bundled collections of related tools)           │  │
│  │  ├── SQLDatabaseToolkit          → Query & inspect DBs     │  │
│  │  ├── PlayWrightBrowserToolkit    → Web scraping/automation │  │
│  │  ├── GitHub / Jira Toolkits      → Issue & repo actions    │  │
│  │  └── Office365 / Gmail Toolkits  → Messaging & calendars   │  │
│  └────────────────────────────────────────────────────────────┘  │
└──────────────────────────────────────────────────────────────────┘

"""
```

### Example of Build in tools

- install:
    - pip install -U duckduckgo-search
    - pip install -U duckduckgo-search langchain-community
    - pip install -U ddgs

```python
# Build in Tools -> DuckDuckGoSearchRun
from langchain_community.tools import DuckDuckGoSearchRun

search_tool = DuckDuckGoSearchRun()

result = search_tool.invoke(
    "latest news in the United States"
)

print(result)

print(result)
```

```python
from langchain_community.tools import DuckDuckGoSearchResults

search_tool = DuckDuckGoSearchResults(
    output_format="list"
)

results = search_tool.invoke(
    "artificial intelligence news"
)

for result in results:
    print(result)
```

```python
# Example: Build-in Tolls : Shell Tools
# pip install langchain-experimental
from langchain_community.tools import ShellTool

shell_tool = ShellTool()

result = shell_tool.invoke("dir")

print(result)
```

#### 1. Create a directory — mkdir

```python
# Create a directory — mkdir
from langchain_community.tools import ShellTool

shell_tools = ShellTool()

# create dir
make_dir = shell_tools.invoke("mkdir langchain_test")

print(make_dir)
```

```python
# 2. Check the directory — dir
result = shell_tool.invoke("dir")

print(result)
```

```python
# 3. Move into the directory — cd
shell_tool.invoke("cd langchain_test")
shell_tool.invoke("dir")
```

```python
# To see the Present working directory
import os
print(os.getcwd())
```

### 4. Create a file

```python
result = shell_tool.invoke(
    "cd langchain_test && type nul > example.txt"
)

print(result)
```

#### 5. Create a file with content

```python
result = shell_tool.invoke(
    'cd langchain_test && echo Hello from LangChain > hello.txt'
)

print(result)
```

```python
# Fetch the data
result = shell_tool.invoke(
    "cd langchain_test && type hello.txt"
)

print(result)
```

#### 6. Delete a file — del

```python
result = shell_tool.invoke(
    "cd langchain_test && del example.txt"
)

print(result)
```

```python
result = shell_tool.invoke(
    "cd langchain_test && dir"
)

print(result)
```

```python
"""     How Tools Work Internally (The Tool-Calling Lifecycle)

┌─────────────────────────────────────────────────────────────┐
│                 TOOL-CALLING INTERNAL FLOW                  │
│                                                             │
│  1. SCHEMA REGISTRATION                                     │
│     Tool definition converted to JSON Schema:               │
│     {                                                       │
│       "name": "search_database",                            │
│       "description": "Searches employee records",           │
│       "parameters": { ... }                                 │
│     }                                                       │
│                                                             │
│  2. BIND TO MODEL                                           │
│     Tool schemas sent inside the API request payload        │
│     to the LLM (e.g., via `tools` parameter in OpenAI)      │
│                                                             │
│  3. LLM DECISION                                            │
│     LLM analyzes user query + tool descriptions.            │
│     Decides: "I need to call `search_database`"             │
│                                                             │
│  4. TOOL CALL EMISSION (JSON)                               │
│     LLM returns an AIMessage with `tool_calls`:             │
│     [{"name": "search_database", "args": {"id": 101}}]      │
│                                                             │
│  5. RUNTIME EXECUTION                                       │
│     LangChain extracts tool name and args, calls Python:    │
│     result = search_database.invoke({"id": 101})            │
│                                                             │
│  6. RETURN CONTEXT TO LLM                                   │
│     Tool output packaged into a `ToolMessage`               │
│                                                             │
│  7. FINAL SYNTHESIS                                         │
│     LLM reads original query + ToolMessage → outputs text   │
└─────────────────────────────────────────────────────────────┘

"""
```
