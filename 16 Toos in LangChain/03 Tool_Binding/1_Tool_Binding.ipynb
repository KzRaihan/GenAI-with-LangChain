### 1. Introduction of Tool Binding

- Definition:
    - Tool Binding is the step where we **register tools** with a Language Model(LLM) so that
    
    - 1. The LLM Knows what tools are available.
    - 2. It knows what each tools does (via description)
    - 3. It knows what input format to use (via schema)


- Tool Binding is the process of connecting one or more tools to an LLM so that the model knows which tools are available and how to call them.


- KEY NOTE:
    - Tool binding gives the model access to tools and their schemas; it does not itself execute the tool.

    - tool binding = connection tools with LLM

```python
""" 
User
   ↓
LLM
   │
   ├── knows → Calculator Tool
   │
   └── knows → Search Tool
             ↓
       Select appropriate tool
             ↓
          Tool Call
             ↓
          Tool Result
             ↓
            LLM
             ↓
        Final Answer

"""
```

### 2. Practical Example 

```python
# Practical Example 1 — Custom Tool Binding

# Step 1 — Create a tool
from langchain_core.tools import tool

@tool
def multiply(a: int, b: int) -> int:
    """Multiply two numbers."""
    return a * b


# Step 2 — Create an LLM
import os
from dotenv import load_dotenv
from langchain_groq import ChatGroq

load_dotenv()

model = ChatGroq(
    model="openai/gpt-oss-20b",
    temperature=0
)

# Step 3 — Bind the tool
model_with_tools = model.bind_tools(
    [multiply] # Bind one and more tool
)

# Step 4 — Give the model a request
response = model_with_tools.invoke(
    "What is 25 multiplied by 10?"
)

print(response)
```

- Key note:
    - The important thing is that the model may return a tool call rather than directly returning the final answer.

```python
print(response.tool_calls)
```

### KeyTakeaway

- Very Important: Binding ≠ Calling

- Binding tells the model:
    - This tool is available, and here is its schema.


#### Practical Example 2 — Multiple Tools

```python
# Step 1 — Create a tool
from langchain_core.tools import tool

# tool 1
@tool
def addition(a:int, b:int)-> int:
    """
    addition of two integer number
    """
    return a + b


# tool 2
@tool
def multiply(a:int, b:int)-> int:
    """
    multiply two number
    """
    return a * b


# Step 2 — Create a LLM
load_dotenv()

model = ChatGroq(
    model="openai/gpt-oss-20b",
    temperature=0.5,
)

# Step 3 — Bind the tool
model_with_tools = model.bind_tools(
    [addition, multiply]
)

# Step 4 — Give the model a request
response_with_tool = model_with_tools.invoke(
    "What is 25 addition 10?"
    )


print(response_with_tool)
```

```python
print(response_with_tool)
```

### Practical Example 3 — Tool Binding + Manual Execution

```python
from langchain_core.tools import tool
from langchain_groq import ChatGroq
from dotenv import load_dotenv

load_dotenv()


# --------------------------------------------------
# 1. Create a custom tool
# --------------------------------------------------

@tool
def multiply(a: int, b: int) -> int:
    """Multiply two numbers."""
    return a * b


# --------------------------------------------------
# 2. Create the LLM
# --------------------------------------------------

model = ChatGroq(
    model="openai/gpt-oss-20b",
    temperature=0
)


# --------------------------------------------------
# 3. Bind the tool to the model
# --------------------------------------------------

model_with_tools = model.bind_tools(
    [multiply]
)


# --------------------------------------------------
# 4. Send a question to the model
# --------------------------------------------------

response = model_with_tools.invoke(
    "What is 25 multiplied by 10?"
)


# --------------------------------------------------
# 5. Inspect the model's tool call
# --------------------------------------------------

print("Tool calls:")
print(response.tool_calls)
```

```python
"""
Tool
  ↓
Bind Tool
  ↓
LLM knows the tool
  ↓
LLM decides to call tool
  ↓
Tool executes
  ↓
Result goes back to LLM
  ↓
Final Answer

"""
```
