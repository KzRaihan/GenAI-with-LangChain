## Topic: ToolKits in LangChain

### 1. Introduction of ToolKits

- Definition:
    - A toolkit is just a collection (bundle) of related tools that serve a common purpose — packaged together for convenience and **reusability**.


- In LangChain:
    - • A toolkit might be: GoogleDriveToolKit
    - • And it can contain the following tools


- Example:
    - Upload a file: GoogleDriveCreateFileTool
    - Search for a file by name/content : GoogleDriveSearchTool
    - Read contents of a file: GoogleDriveReadFileTool

```python
# Example 1 : ToolKits (combine relevant two tool)
from langchain_core.tools import tool

# Custom tools 1
@tool
def add(a: int, b: int) -> int:
    """Add two numbers"""
    return a + b


# Custom tools 2
@tool
def multiply(a: int, b: int) -> int:
    """Multiply two numbers"""
    return a * b

# Create a class to combine two tools
class MathToolkit:
    def get_tools(self):
        return [add, multiply]



# object of toolkit 
toolkit = MathToolkit()
tools = toolkit.get_tools()

for tool in tools:
    print(tool.name, "=>", tool.description)

```
