### 1. Introduction of json outpur parser
- Definition:
    - The JSON Output Parser is used when we want the LLM to return data in JSON format.

```python
import os
from dotenv import load_dotenv

from langchain_huggingface import ChatHuggingFace, HuggingFaceEndpoint
from langchain_core.prompts import PromptTemplate
from langchain_core.output_parsers import JsonOutputParser

# ------------------------------------------------------------------
# STEP 1: Load variables from the .env file
# ------------------------------------------------------------------
load_dotenv()

# STEP 2: Define a working, supported model endpoint
llm = HuggingFaceEndpoint(
    repo_id="TinyLlama/TinyLlama-1.1B-Chat-v1.0",
    task="text-generation",
    temperature=0.1,  # Lower temperature helps strictly follow JSON schema
)

# STEP 3: Instantiate the Chat Model (FIXED)
model = ChatHuggingFace(llm=llm)

# STEP 4: Initialize the JsonOutputParser object
parser = JsonOutputParser()

# STEP 5: Define the schema of prompt
template = PromptTemplate(
    template="Give me the name, age and city of a fictional person \n {format_instruction}",
    input_variables=[],
    partial_variables={"format_instruction": parser.get_format_instructions()},
)

# STEP 6: Format, invoke, and parse
prompt = template.format()

# get the model response
response = model.invoke(prompt)

# parser the model response
final_response = parser.parse(response.content)

print(final_response)
print(type(final_response))
print(final_response['name'])
```

```python
# use chain component

import os
from dotenv import load_dotenv

from langchain_huggingface import ChatHuggingFace, HuggingFaceEndpoint
from langchain_core.prompts import PromptTemplate
from langchain_core.output_parsers import JsonOutputParser

# ------------------------------------------------------------------
# STEP 1: Load variables from the .env file
# ------------------------------------------------------------------
load_dotenv()

# STEP 2: Define a working, supported model endpoint
llm = HuggingFaceEndpoint(
    repo_id="TinyLlama/TinyLlama-1.1B-Chat-v1.0",
    task="text-generation",
    temperature=0.1,  # Lower temperature helps strictly follow JSON schema
)

# STEP 3: Instantiate the Chat Model 
model = ChatHuggingFace(llm=llm)

# STEP 4: Initialize the JsonOutputParser object
parser = JsonOutputParser()

# STEP 5: Define the schema of prompt
template = PromptTemplate(
    template="Give me 5 facts about {topic} \n {format_instruction}",
    input_variables=["topic"],
    partial_variables={"format_instruction": parser.get_format_instructions()},
)

# define chain
chain = template | model | parser 

response = chain.invoke({
    "topic": "Generative AI"
}) 

print(response)
```

- Key note:
    - the response(json) of the model output is not actual well-structure because here, the model is define the output structure (json)
