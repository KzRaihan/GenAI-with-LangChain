### 1. Introduction of TextLoader
- Definition:
    - TextLoader is a simple and commonly used document loader in LangChain that reads plain text(.txt) files and converts them into LangChain Document objects.
    

### 2. Use case and Limitation

- 1. Use case:
    - Ideal for loading chat logs, scraped text, transcripts, code snippets, or any plain text data.


- 2. Limitation:
    - Works only with .txt files

### 3. Practical Example of TextLoader

```python
# Example 1: TextLoader

from langchain_community.document_loaders import TextLoader

# Create an object of TextLoader
loader = TextLoader("cricket.txt", encoding="utf-8") # path of txt file

# load the Text file into document(entire text files)
document = loader.load()

# print 
print(document)
print(type(document))
```

- Key Note:
    - Every Loader(TextLoader, ..) load the document into python list format.

    


```python
# Total number of document 
print(len(document))
```

- Key Note:
    - for this case, there is only one document or one txt file 

```python
# Extract the first document 
print(document[0])
print(type(document[0]))
```

- Key Note:
    - Each document has two key
        - 1. page_content
            - The Actual text context
        
        - 2. metadata
            - source: filename.pdf, others

```python
# if we want to extract only the page_context 
print(document[0].page_content)
```

```python
# if we want to extract only the metadata 
print(document[0].metadata)
```

### TextLoader with model integration

```python
# Example 2: TextLoader + model integration

from langchain_community.document_loaders import TextLoader
from langchain_openai import ChatOpenAI
from langchain_core.prompts import PromptTemplate, ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser
from dotenv import load_dotenv

load_dotenv()

# Create an object of TextLoader
loader = TextLoader("cricket.txt", encoding="utf-8") # path of txt file

# load the Text file into document(entire text files)
document = loader.load()

# Define the model
model = ChatOpenAI()

# Define the Prompt
prompt = ChatPromptTemplate.from_template(
    "Summarize the following poem {poem}"
)

# Define parser
parser = StrOutputParser()

# Create a RunnableSequential Chain
chain = prompt | model | parser


# invoke the model
response = model.invoke(
    {
        "poem": document[0].page_content
    }
)

print(f"The Actual Poem: \n{document[0].page_content}")

print(f"Summarize of the Poem: \n {response}")
```

```python
# Example- 3: Convert the Document into Runnable

# Purpose: TextLoader -> docs -> Runnable docs -> model integration

from langchain_community.document_loaders import TextLoader
from langchain_openai import ChatOpenAI
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser
from langchain_core.runnables import RunnableLambda
from dotenv import load_dotenv

load_dotenv()

# Create an object of TextLoader
loader = TextLoader("cricket.txt", encoding="utf-8")

# Load the text file into documents
docs = loader.load()

# Convert docs into runnable
docs_runnable = RunnableLambda(lambda x: {"poem": docs[0].page_content})

# Define the model
model = ChatOpenAI()

# Define the prompt
prompt = ChatPromptTemplate.from_template(
    "Summarize the following poem:\n\n{poem}"
)

# Define parser
parser = StrOutputParser()

# Create chain
chain = docs_runnable | prompt | model | parser

# Invoke the chain
response = chain.invoke({})

print(f"The Actual Poem:\n{docs[0].page_content}")

print(f"\nSummary of the Poem:\n{response}")
```
