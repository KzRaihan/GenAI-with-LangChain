### 3. Practical Example of CSVLoader

```python
from langchain_community.document_loaders import CSVLoader

FILE_PATH = "Social_Network_Ads.csv"

loader = CSVLoader(FILE_PATH)

# Load all documents
documents = loader.load()

print(f"Length of file: {len(documents)}")

print(f"\n\nFirst row content:\n\n{documents[0].page_content}")
```
