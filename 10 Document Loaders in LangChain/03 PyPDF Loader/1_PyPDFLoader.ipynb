### 1. Introduction of PyPDFLoader

- Definition:
    - PyPDFLoader is a document loader in LangChain used to load content from PDF files and convert each page into a Document object.


    - PyPDFLoader is a LangChain document loader that uses the pypdf library under the hood to read PDF files and convert each page into a separate LangChain Document object.



```python
"""     - How It Works Internally?
┌─────────────────────────────────────────────────────────────┐
│           PyPDFLoader INTERNAL FLOW                         │
│                                                             │
│  1. RECEIVE FILE PATH OR URL                                │
│     loader = PyPDFLoader("report.pdf")                      │
│                                                             │
│  2. OPEN THE PDF FILE                                       │
│     Internally calls: pypdf.PdfReader("report.pdf")         │
│     This reads the PDF structure (pages, metadata, fonts)   │
│                                                             │
│  3. EXTRACT TEXT PAGE BY PAGE                               │
│     For each page in the PDF:                               │
│       page = reader.pages[i]                                │
│       text = page.extract_text()  ← pypdf does the work     │
│                                                             │
│  4. BUILD DOCUMENT OBJECTS                                  │
│     For each page's text:                                   │
│       Document(                                             │
│           page_content="extracted text from page N",        │
│           metadata={                                        │
│               "source": "report.pdf",                       │
│               "page": N  (0-indexed)                        │
│           }                                                 │
│       )                                                     │
│                                                             │
│  5. RETURN List[Document]                                   │
│     [Doc(page=0), Doc(page=1), Doc(page=2), ...]            │
│     One Document per page!                                  │
└─────────────────────────────────────────────────────────────┘


"""
```

### 2. Use case and Limitation


```python
""" 
┌─────────────────────────────────────────────────────────────┐
│              WHY PyPDFLoader? / USE CASE                    │
├─────────────────────────────────────────────────────────────┤
│                                                             │
│ -  ZERO system dependencies                                 │
│     Pure Python — no need to install poppler, tesseract,    │
│     or any OS-level packages. Just pip install pypdf.       │
│                                                             │
│ -  FAST                                                     │
│     Extracts text from a 100-page PDF in < 1 second.        │
│                                                             │
│ -  LIGHTWEIGHT                                              │
│     pypdf is ~2MB. Compare to unstructured (~500MB).        │
│                                                             │
│ -  PAGE-LEVEL GRANULARITY                                   │
│     Each page becomes a separate Document with page number  │
│     in metadata — perfect for citation in RAG.              │
│                                                             │
│ -  URL SUPPORT                                              │
│     Can load PDFs directly from URLs, not just local files. │
│                                                             │
│ -  PASSWORD-PROTECTED PDF SUPPORT                           │
│     Can open encrypted PDFs with a password parameter.      │
│                                                             │
│ -  MOST POPULAR                                             │
│     The default choice in 90% of LangChain RAG tutorials.   │
└─────────────────────────────────────────────────────────────┘

"""
```

```python
""" 
┌─────────────────────────────────────────────────────────────┐
│              LIMITATIONS                                    │
├─────────────────────────────────────────────────────────────┤
│                                                             │
│ -  NO OCR (Optical Character Recognition)                   │
│     Scanned PDFs (images of text) return empty strings.     │
│     → Use: PyMuPDFLoader or UnstructuredPDFLoader + OCR     │
│                                                             │
│ -  POOR TABLE EXTRACTION                                    │
│     Complex tables get mangled into unstructured text.      │
│     → Use: PDFPlumberLoader for table-heavy PDFs            │
│                                                             │
│ -  NO IMAGE EXTRACTION                                      │
│     Charts, diagrams, and figures are ignored.              │
│     → Use: UnstructuredFileLoader with hi_res strategy      │
│                                                             │
│ -  FORMATTING LOSS                                          │
│     Bold, italic, headers, and columns are flattened.       │
│     → Use: PDFMinerLoader for layout-aware extraction       │
└─────────────────────────────────────────────────────────────┘

"""
```

```python
""" 
What kind of PDF do you have?
│
├─ Text-based (reports, articles, contracts)
│   └─→ PyPDFLoader  (fast, lightweight)
│
├─ Table-heavy (financial reports, data sheets, table)
│   └─→ PDFPlumberLoader  (best table extraction)
│
├─ Scanned / Image-based (old documents, receipts)
│   └─→ UnstructuredFileLoader + OCR 
│
├─ Mixed (text + images + tables)
│   └─→ UnstructuredFileLoader  (handles everything)
│
└─ Need maximum speed
    └─→ PyMuPD
    
- LangChain Document link:
    - https://docs.langchain.com/oss/python/integrations/document_loaders

"""
```

```python
""" 


### One-line definitions
 
- 1. PyPDFLoader
    - Converts PDF content into LangChain Document objects.

- 2. Document
    - Stores extracted content (page_content) and associated metadata.

- 3. Text Splitter
    - Divides documents into smaller chunks suitable for embedding and retrieval.

- 4. Embedding Model
    - Converts chunks into numerical vectors.

- 5. Vector Database
    - Stores and indexes those vectors for similarity search.

- 6. Retriever
    - Finds the chunks most relevant to a user's query.

"""
```

###  3. Practical Example of PyPDFLoader

```python
# Example 1: load a pdf
from langchain_community.document_loaders import PyPDFLoader

# create an object of PyPDFLoader
loader = PyPDFLoader("dl-curriculum.pdf") # path of pdf

# convert to docs
docs = loader.load()

print(docs)
```

### 4. Complete Summary

```python
""" 
┌──────────────────────────────────────────────────────────────────┐
│                       PyPDFLoader                                │
│                                                                  │
│  WHAT:  Loads PDF files into List[Document] (one per page)       │
│  LIB:   pypdf (pure Python, ~2MB, no system deps)                │
│  SPEED:  Fast (100 pages < 1 second)                             │
│                                                                  │
│  IMPORT:                                                         │
│    from langchain_community.document_loaders import PyPDFLoader  │
│                                                                  │
│  BASIC USAGE:                                                    │
│    loader = PyPDFLoader("report.pdf")                            │
│    docs = loader.load()           # All pages at once            │
│    docs = loader.lazy_load()      # One page at a time           │
│                                                                  │
│  OUTPUT:                                                         │
│    Document(                                                     │
│        page_content="text from page N",                          │
│        metadata={"source": "report.pdf", "page": N}  (0-indexed) │
│    )                                                             │
│                                                                  │
│  FEATURES:                                                       │
│     - Local files and URLs                                       │
│     - Password-protected PDFs                                    │
│     - Page-level granularity                                     │
│     - Zero system dependencies                                   │
│     - No OCR (scanned PDFs)                                      │
│     - Poor table extraction                                      │
│     - No image extraction                                        │
│                                                                  │
│  ALTERNATIVES:                                                   │
│    Tables → PDFPlumberLoader                                     │
│    Scanned → UnstructuredFileLoader + OCR                        │
│    Speed → PyMuPDFLoader                                         │
│    Mixed → UnstructuredFileLoader                                │
│                                                                  │
│  PIPELINE:                                                       │
│    PyPDFLoader → TextSplitter → Embeddings → VectorStore → RAG  │
│                                                                  │
│  GOLDEN RULE:                                                    │
│  "Use PyPDFLoader for text-based PDFs. Switch to PDFPlumber      │
│   for tables or Unstructured for scanned/complex documents."     │
└──────────────────────────────────────────────────────────────────┘

"""
```
