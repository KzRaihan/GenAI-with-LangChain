### 1. Introduction of WebBaseLoader

- Definition:
    - WebBaseLoader is a document loader in LangChain used to load and extract text context from web pages(URLs).

    - It uses request and BeautifulSoup under the hood to parse HTML and extract visible text.

### 2. Use and Limitation

- 1. Use:
    - For blogs, news articles, or public websites where the content is primarily tex-based and static.

- 2. Limitation:
    - Doesn't handle JavaScript-heavy pages well (use SeleniumURLLoader for that).
    - Loads only static context(what's in the HTML, not what loads after the page renders.)

### 3. Practical Example of WebBaseLoader

```python
from langchain_community.document_loaders import WebBaseLoader

FILE_PATH = "https://www.rokomari.com/book/215765/englishe-durbolder-jonno-english-therapy"

loader = WebBaseLoader(FILE_PATH)

# Load all documents
documents = loader.load()

print(len(documents))

print(documents[0].page_content)
```

- Key Note:
    - Always return the single or one document.
    - also we can pass the multiple  url or file path as list format so that we access the multiple content of each url.

```python
from langchain_community.document_loaders import WebBaseLoader
from langchain_openai import ChatOpenAI
from langchain_core.output_parsers import StrOutputParser
from langchain_core.prompts import PromptTemplate
from dotenv import load_dotenv

load_dotenv()

model = ChatOpenAI()

prompt = PromptTemplate(
    template='Answer the following question \n {question} from the following text - \n {text}',
    input_variables=['question','text']
)

parser = StrOutputParser()

url = "https://www.rokomari.com/book/215765/englishe-durbolder-jonno-english-therapy"

# load the text
loader = WebBaseLoader(url)


docs = loader.load()

# define the chain
chain = prompt | model | parser

response = chain.invoke(
    {'question':'What is the product that we are talking about?', 'text':docs[0].page_content}
    )

print(response)
```
