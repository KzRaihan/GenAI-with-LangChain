## Topic: TypedDict 

### 1. Introduction of TypedDict 

- Definition:
    - TypedDict is a way to define a dictionary in Python where we specify what keys and values should exist. 

    - It helps ensure that our dictionary follows a specific structure.


### 2. Why use TypedDict?
- Use: 
    -  It tells Python what keys are required and what types of values they should have.

    -  It does not validate data at runtime (it just helps with type hints for better coding).



- Key reasons to use: 
    - Defines the expected keys and value types of a dictionary.

    - Improves code readability and type checking.

    - Useful when structured data should remain a normal Python dictionary rather than a Pydantic object.


- for use TypedDict we use 
    - from typing import TypedDict


```python
### Example 1: 
from typing import TypedDict


# ---------------------------------------------------------
# Define the expected dictionary structure
# ---------------------------------------------------------

class Student(TypedDict):

    # Student's name must be a string
    name: str

    # Student's age must be an integer
    age: int

    # Student's GPA must be a float
    gpa: float


# ---------------------------------------------------------
# Create a dictionary following the structure of Student Class
# ---------------------------------------------------------

student: Student = {
    "name": "Rahim",
    "age": 22,
    "gpa": 3.85
}


# ---------------------------------------------------------
# Access the dictionary values
# ---------------------------------------------------------

print(student["name"])
print(student["age"])
print(student["gpa"])

print(type(student["gpa"]))
```

```python
### Example 2: TypedDict with LangChain Structured Output

from langchain_openai import ChatOpenAI
from dotenv import load_dotenv

from typing import TypedDict

# load the openai api from .env file
load_dotenv()

# ---------------------------------------------------------
# 1. Create the chat model
# ---------------------------------------------------------

model = ChatOpenAI(
    model="gpt-4o"
)


# ---------------------------------------------------------
# 2. Define the structured output schema
# ---------------------------------------------------------
class Review(TypedDict):
    # summary -> string format
    summary : str

    # sentiment -> string format
    sentiment : str

# ---------------------------------------------------------
# 3. Configure structured output that follow the Review class
# ---------------------------------------------------------
structured_model = model.with_structured_output(Review)


# ---------------------------------------------------------
# 4. Invoke the structured_model
# ---------------------------------------------------------
response = structured_model.invoke(
    """ 
    I recently upgraded to the Samsung Galaxy S24 Ultra, and I must say, it’s an absolute powerhouse! The Snapdragon 8 Gen 3 processor makes everything lightning fast—whether I’m gaming, multitasking, or editing photos. The 5000mAh battery easily lasts a full day even with heavy use, and the 45W fast charging is a lifesaver.
    
    """
)

# ---------------------------------------------------------
# 5. Display the structured response
# ---------------------------------------------------------

print(response)

print(type(response))

print(response['summary'])
print(response['sentiment'])
```

### TypedDict — Key Pros
- 1. Simple and lightweight — 
    - Provides a clear dictionary structure without creating complex model classes.

- 2. Improves type safety — 
    - Defines expected keys and value types, helping IDEs and type checkers detect mistakes.

- 3. Easy dictionary access — 
    - Data can be accessed using familiar syntax like student["name"].

- 4. Works well with LangChain — 
    - Can be used as a schema for structured LLM output.

### TypedDict — Key Cons
- 1. No built-in runtime validation — 

    - TypedDict mainly provides static type information; it does not validate data like Pydantic.

- 2. Less powerful than Pydantic — 
    - It doesn't provide features such as Field() descriptions and advanced validation.

- 3. Errors may appear only during type checking —
    - Incorrect data can still exist at runtime unless another validation mechanism is used.

- 4. Less suitable for complex data validation — 
    - For production applications requiring strong validation, Pydantic is generally a better choice.

- 5.Requires correct type-checking tools — 
    - The benefits of type checking are strongest when using tools such as Pyright or MyPy.

### Summary
- TypedDict is a Python typing feature used to define the expected keys and value types of a dictionary. In LangChain Structured Output, it can be used as a lightweight schema when we want the LLM to return structured dictionary-like data.

- It is only use for data representation or type hint.

- It can't provide the Data Validation Functionality.
