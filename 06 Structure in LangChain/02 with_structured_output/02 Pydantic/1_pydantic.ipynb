## Topic: Pydantic in  LangChain Structured Output and Data Validation

### 1. Introduction of Pydantic
- Definition:
    - Pydantic is a Python library used to define structured data models and validate data using Python type hints.

    
- Why do we need pydantic?
    - To ensure that data follows a predefined structure and satisfies expected data types.

    - It ensures that the data we work with is correct, structured, and type-safe.



```python
# What Problem Does Pydantic Solve?

from typing import TypedDict

# ---------------------------------------------------------
# Define the expected dictionary structure
# ---------------------------------------------------------

class Student(TypedDict):

    # Student's name must be a string
    name: str

    # Student's age must be an integer
    age: int

    # Student's GPA must be a float
    gpa: float


# ---------------------------------------------------------
# Create a dictionary following the structure of Student Class
# ---------------------------------------------------------

student: Student = {
    "name": "Rahim",
    "age": 22,
    "gpa": "Three point Eight-Five"
}


# ---------------------------------------------------------
# Access the dictionary values
# ---------------------------------------------------------

print(student["name"])
print(student["age"])
print(student["gpa"])

print(type(student["gpa"]))
```

- Key Note:
    - Here, define into expected dictionary structure the gpa is float but we pass into string. for this our program can't raise any error.


- Python's normal dictionary doesn't automatically enforce our expected structure.

- Here is Come Pydantic to validate the data type.

- For use pydantic we use
    - from pydantic import BaseModel

```python
## Example 1: Base Example
from pydantic import BaseModel


# ---------------------------------------------------------
# Define a Pydantic model
# ---------------------------------------------------------
class Student(BaseModel):
    name: str
    age : int


# ---------------------------------------------------------
# Create a object following the structure of Student Class
# ---------------------------------------------------------
student = Student(
        name = "KzRaihan",
        age = 23,
    )


# ---------------------------------------------------------
# Access the values
# ---------------------------------------------------------
print(f"Student Name: {student.name}")
print(f"Student Age : {student.age}")
```

## Topic: Field Validation 
- Field:
    - Used to add extra rules, constraints, and metadata to a model field. Controls validation like min/max values, length, default values, and descriptions.

    - One-line: 
        - Adds validation rules and metadata to a field.



```python
# Example: 4: Pydantic Field 

# import necessary library
from pydantic import BaseModel, EmailStr, Field

from typing import List, Dict, Optional, Annotated

# Define the Schema
class Profile(BaseModel):
    # define schema
    name : str = Field(min_length= 2, max_length= 50)

    age  : int = Field(gt= 0, le= 100, description="age must be greater than zero and less than 100") # ge -> Greater than or le : less than

    weight: float = Field(gt= 0, description="weight value must be greater than zero")

    married: Optional[bool] = Field(default=None, description = "either True or False")

    email : EmailStr

    contact_details: Annotated[Dict,Field(default=None, description='Give your email Address')]


# ---------------------------------------------------------
# Create a object following the structure of Student Class
# ---------------------------------------------------------
student = Profile(
    name = "KzRaihan",
    age = 23,
    weight = 75.55,
    email = "kzraihan@gmai.com",
    contact_details = {
        "mobile" : 5555555,
        "address": "Dhaka"
    }


)

# ---------------------------------------------------------
# Access the values into correct type way
# ---------------------------------------------------------
print(f"Student Name: {student.name}")
print(f"Student age: {student.age}")
print(f"Email: {student.email}")
print(f"Contact information: {student.contact_details}")
```

### LangChain + Pydantic Example

```python
# LangChain + Pydantic Example
from pydantic import BaseModel, Field
from langchain_openai import ChatOpenAI


# ---------------------------------------------------------
# 1. Define the output schema
# ---------------------------------------------------------

class Student(BaseModel):

    name: str = Field(
        description="The student's full name"
    )

    degree: str = Field(
        description="The student's academic degree"
    )

    gpa: float = Field(
        description="The student's GPA"
    )


# ---------------------------------------------------------
# 2. Create the chat model
# ---------------------------------------------------------

model = ChatOpenAI(
    model="gpt-4o"
)


# ---------------------------------------------------------
# 3. Configure structured output
# ---------------------------------------------------------

structured_model = model.with_structured_output(
    Student
)


# ---------------------------------------------------------
# 4. Send input to the model
# ---------------------------------------------------------

response = structured_model.invoke(
    """
    Extract the student information.

    Md Kamruzzaman is a Computer Science and Engineering
    graduate with a GPA of 3.85.
    """
)


# ---------------------------------------------------------
# 5. Access structured fields
# ---------------------------------------------------------

print("Name:", response.name)
print("Degree:", response.degree)
print("GPA:", response.gpa)
```

```python
# Example: Pydantic with langchain
# ===================================================

# import Necessary libarary
from langchain_openai import ChatOpenAI
from dotenv import load_dotenv
from pydantic import BaseModel, Field, EmailStr
from typing import Dict, List, Optional, Annotated

load_dotenv()

# define the model
model = ChatOpenAI(
   model="gpt-4o"
)

# Define the schema
class Profile(BaseModel):
    name : str = Field(
        description = "Give your Full name"
    )

    age : int = Field(
        gt = 0, le = 100,
        description="age is greater than zero and less than 100"
    )

    email: EmailStr  # Require Field

    weight : Optional[float] = Field(
        gt = 0, le = 100,
        description="Weight is Optional but in between 1-99"
    )

    contact_details: Annotated[Dict, Field(default=None, description="Give Your Contact Details")]


# Structure output
structured_model = model.with_structured_output(Profile)

# model invoke
response = structured_model.invoke(
"""  
Rahim is a student. He is 23 year old and he is a brilliant student. he email is rahim22@gmai.com. His phone number is 00012451


"""
)

# ---------------------------------------------------------
# 5. Access structured fields
# ---------------------------------------------------------

print("Name:", response.name)
print("Degree:", response.degree)
print("GPA:", response.gpa)
```

```python
from langchain_openai import ChatOpenAI
from dotenv import load_dotenv
from typing import TypedDict, Annotated, Optional, Literal
from pydantic import BaseModel, Field

load_dotenv()

model = ChatOpenAI()

# schema
class Review(BaseModel):

    key_themes: list[str] = Field(description="Write down all the key themes discussed in the review in a list")
    summary: str = Field(description="A brief summary of the review")
    sentiment: Literal["pos", "neg"] = Field(description="Return sentiment of the review either negative, positive or neutral")
    pros: Optional[list[str]] = Field(default=None, description="Write down all the pros inside a list")
    cons: Optional[list[str]] = Field(default=None, description="Write down all the cons inside a list")
    name: Optional[str] = Field(default=None, description="Write the name of the reviewer")
    

structured_model = model.with_structured_output(Review)

result = structured_model.invoke("""I recently upgraded to the Samsung Galaxy S24 Ultra, and I must say, it’s an absolute powerhouse! The Snapdragon 8 Gen 3 processor makes everything lightning fast—whether I’m gaming, multitasking, or editing photos. The 5000mAh battery easily lasts a full day even with heavy use, and the 45W fast charging is a lifesaver.

The S-Pen integration is a great touch for note-taking and quick sketches, though I don't use it often. What really blew me away is the 200MP camera—the night mode is stunning, capturing crisp, vibrant images even in low light. Zooming up to 100x actually works well for distant objects, but anything beyond 30x loses quality.

However, the weight and size make it a bit uncomfortable for one-handed use. Also, Samsung’s One UI still comes with bloatware—why do I need five different Samsung apps for things Google already provides? The $1,300 price tag is also a hard pill to swallow.

Pros:
Insanely powerful processor (great for gaming and productivity)
Stunning 200MP camera with incredible zoom capabilities
Long battery life with fast charging
S-Pen support is unique and useful
                                 
Review by Nitish Singh
""")

print(result)
```

### Key Takeaway

- BaseModel:
    - The foundation class of Pydantic. Every model we create inherits from it. It provides automatic data validation, type checking, and conversion.

    - One-line: 
        - Parent class(BaseModel) that gives validation superpowers to our models.

- EmailStr:
    - A special string type that validates email format automatically. Rejects invalid emails like "abc" and accepts valid ones like "user@mail.com".

    - One-line: 
        - A string that must be a valid email address.


- AnyUrl
    - A special type that validates URL format. Ensures the value is a proper URL with scheme (http, https, ftp, etc.) and domain.

    - One-line: 
        - A string that must be a valid URL.


- Field:
    - Used to add extra rules, constraints, and metadata to a model field. Controls validation like min/max values, length, default values, and descriptions.

    - One-line: 
        - Adds validation rules and metadata to a field.

- field_validator:
    - A decorator used to write custom validation logic for a field. When built-in validation is not enough, we write our own rules.

    - One-line: 
        - Decorator for writing our own custom validation rules.
