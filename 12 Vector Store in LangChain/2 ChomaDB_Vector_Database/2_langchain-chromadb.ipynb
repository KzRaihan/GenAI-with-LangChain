#### CREATE - ChromaDB 

```python
# --------------------------------------------------
# 1. Import Embedding model + Chat model + Chroma
# --------------------------------------------------
from langchain_huggingface import HuggingFaceEmbeddings
from langchain_groq import ChatGroq
from langchain.vectorstores import Chroma


# --------------------------------------------------
# 1. Load environment variables
# --------------------------------------------------
import os
from dotenv import load_dotenv, find_dotenv
_ = load_dotenv(find_dotenv())


# -----------------------------------------------------------------------------
# 2. Create embedding model (Download the HuggingFaceEmbeddings model)
# -----------------------------------------------------------------------------
embeddings = HuggingFaceEmbeddings(
    model_name="sentence-transformers/all-MiniLM-L6-v2"
)

# --------------------------------------------------
# 3. Create Actual model
# --------------------------------------------------
llamaChatModel = ChatGroq(
    model="openai/gpt-oss-safeguard-20b"
)

# --------------------------------------------------
# 4. Create Chroma vector store
# --------------------------------------------------
vector_store = Chroma(
    collection_name="genai_demo",
    embedding_function=embeddings,
    persist_directory="./chroma_db"
)

```

```python
# --------------------------------------------------
# 5. CREATE
# Add documents to ChromaDB
# --------------------------------------------------
from langchain_core.documents import Document

# Create LangChain documents for IPL players

doc1 = Document(
        page_content="Virat Kohli is one of the most successful and consistent batsmen in IPL history. Known for his aggressive batting style and fitness, he has led the Royal Challengers Bangalore in multiple seasons.",
        metadata={"team": "Royal Challengers Bangalore"}
    )
doc2 = Document(
        page_content="Rohit Sharma is the most successful captain in IPL history, leading Mumbai Indians to five titles. He's known for his calm demeanor and ability to play big innings under pressure.",
        metadata={"team": "Mumbai Indians"}
    )
doc3 = Document(
        page_content="MS Dhoni, famously known as Captain Cool, has led Chennai Super Kings to multiple IPL titles. His finishing skills, wicketkeeping, and leadership are legendary.",
        metadata={"team": "Chennai Super Kings"}
    )
doc4 = Document(
        page_content="Jasprit Bumrah is considered one of the best fast bowlers in T20 cricket. Playing for Mumbai Indians, he is known for his yorkers and death-over expertise.",
        metadata={"team": "Mumbai Indians"}
    )
doc5 = Document(
        page_content="Ravindra Jadeja is a dynamic all-rounder who contributes with both bat and ball. Representing Chennai Super Kings, his quick fielding and match-winning performances make him a key player.",
        metadata={"team": "Chennai Super Kings"}
    )

docs = [doc1, doc2, doc3, doc4, doc5]

# Create document with custom ids
ids = vector_store.add_documents(
    documents=docs,
    ids=["doc_001", "doc_002", "doc_003", "doc_004", "doc_005"]
)

print("Created documents:")
print(ids)
```

####  add documents

```python
# add documents
vector_store.add_documents(docs)
```

```python
# view Direct document retrieval by ID or filter
doc = vector_store.get(
    ids=["doc_001"],
    where={"team": "Royal Challengers Bangalore"}   # metadata filter
)

print(doc)
```

```python
# Basic similarity search on database
results = vector_store.similarity_search(
    query='Who among these are a bowler?',
    k=2 # return two relevant records
)

print(f"Response 1 content : {results[0].page_content}")
print(f"\nResponse 1 metadata: {results[0].metadata}")

print("\n\n\n")

print(f"Response 2 content : {results[1].page_content}")
print(f"\nResponse 2 metadata: {results[1].metadata}")
```

```python

# With relevance score or search with similarity score
results_scored = vector_store.similarity_search_with_score(
    query='Who among these are a bowler?',
    k=2
)

# Returns list of (Document, float_score)
print(f"Response 1 content : {results_scored[0]}")

print("\n")

print(f"Response 2 content : {results_scored[1]}")
```

#### update documents

```python
# update documents
updated_doc1 = Document(
    page_content="Virat Kohli, the former captain of Royal Challengers Bangalore (RCB), is renowned for his aggressive leadership and consistent batting performances. He holds the record for the most runs in IPL history, including multiple centuries in a single season. Despite RCB not winning an IPL title under his captaincy, Kohli's passion and fitness set a benchmark for the league. His ability to chase targets and anchor innings has made him one of the most dependable players in T20 cricket.",
    metadata={"team": "Royal Challengers Bangalore"}
)
vector_store.update_document(document_id='7fb88c53-c156-40a0-8102-62885a1e0668',document=updated_doc1)
```

```python
# view the update document retrieval by ID or filter
doc = vector_store.get(
    ids=["doc_001"]
)

print(doc)
```
