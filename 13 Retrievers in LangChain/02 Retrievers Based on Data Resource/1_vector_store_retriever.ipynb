###  1. Introduction of Vector Store retriever

- Definition:
    - A Vector Store Retriever in LangChain is the most common type of retriever that lets we search and fetch documents from a vector store based on semantic similarity using vector embeddings.


```python
""" 
workflow:
=========

[Loader] → [Splitter] → [Embeddings] → [Vector Store]
                                                  ↓
User Query → [Embed Query] → [SIMILARITY SEARCH] ← THIS COMPONENT
                                                  ↓
                                          Top-k Documents
                                                  ↓
                                            [Prompt → LLM]

"""
```

```python
"""
         Introduction of Similarity Search retriever
         ============================================

┌─────────────────────────────────────────────────────────────┐
│           SIMILARITY SEARCH RETRIEVER                       │
│                                                             │
│  WHAT:  Default retriever that finds top-k closest chunks   │
│         to a query using vector distance                    │
│                                                             │
│  WHY:   Fast, simple, and effective semantic search         │
│                                                             │
│  HOW:   1. Embed the query → vector                         │
│         2. Compare with stored document vectors             │
│         3. Return k most similar documents                  │
│                                                             │
│  DISTANCE METRICS:                                          │
│    Cosine Similarity → Angle between vectors (most common)  │
│    Dot Product       → Magnitude + direction                │
│    Euclidean (L2)    → Straight-line distance               │
│                                                             │
│  INTUITION:                                                 │
│    "Meaning close in text = vectors close in space"         │
│                                                             │
│    Query: "AI in medicine"                                  │
│       ↓ (embed)                                             │
│    [0.12, 0.88, 0.34, ...]                                  │
│       ↓ (compare)                                           │
│    Returns docs about "AI healthcare", "medical ML"         │
│                                                             │
│  KEY TRAIT:                                                 │
│    Returns MOST RELEVANT docs (no diversity control)        │
│    → May return near-duplicate/redundant results            │
└─────────────────────────────────────────────────────────────┘




"""
```

### 2. Syntax of Similarity Search retriever

```python
""" 
┌─────────────────────────────────────────────────────────────┐
│        SIMILARITY SEARCH RETRIEVER SYNTAX                   │
│                                                             │
│  METHOD 1: Direct Vector Store Search                       │
│    docs = vectorstore.similarity_search(                    │
│        query="your question",                               │
│        k=4                                                  │
│    )                                                        │
│===========================================================  │
│  METHOD 2: As a Retriever (Recommended)                     │
│    retriever = vectorstore.as_retriever(                    │
│        search_type="similarity",                            │
│        search_kwargs={"k": 4}                               │
│    )                                                        │
│    docs = retriever.invoke("your question")                 │
│===========================================================  │
│                                                             │
│  METHOD 3: With Scores                                      │
│    results = vectorstore.similarity_search_with_score(      │
│        query="your question",                               │
│        k=4                                                  │
│    )                                                        │
│    # returns [(Document, score), ...]                       │
│===========================================================  │
│                                                             │
│  METHOD 4: With Metadata Filter                             │
│    retriever = vectorstore.as_retriever(                    │
│        search_type="similarity",                            │
│        search_kwargs={                                      │
│            "k": 4,                                          │
│            "filter": {"category": "AI"}                     │
│        }                                                    │
│    )                                                        │
│                                                             │
│  KEY PARAMS:                                                │
│    query  → Text to search for                              │
│    k      → Number of docs to return [default: 4]           │
│    filter → Optional metadata restriction                   │
└─────────────────────────────────────────────────────────────┘

- Key NOTE:
    similarity → "Give me the k closest, period."
    mmr → "Give me k results that are relevant AND diverse."
    similarity_score_threshold → "Give me results above a score cutoff."

"""
```

### 3. Practical Examples of Similarity Search retriever with FAISS Database

```python
# import necessary library
from langchain_core.documents import Document
from langchain_huggingface import HuggingFaceEmbeddings
from langchain_community.vectorstores import FAISS
```

```python
# Step 1: Your source documents
documents = [
    Document(page_content="LangChain helps developers build LLM applications easily."),
    Document(page_content="Chroma is a vector database optimized for LLM-based search."),
    Document(page_content="Embeddings convert text into high-dimensional vectors."),
    Document(page_content="OpenAI provides powerful embedding models."),
]
```

```python
# Step 2: Initialize the Embedding model
# Download from the Huggingface platform then initialize the model 

embedding_model = HuggingFaceEmbeddings(
    model_name="sentence-transformers/all-MiniLM-L6-v2"
)
```

```python
# Step 3: Create a FAISS Vector Database
vectorstore = FAISS.from_documents(
    documents = documents,
    embedding= embedding_model
    
)
```

```python
# Step 4: Convert vectorstore into a retriever
retriever = vectorstore.as_retriever(
    search_type = "similarity", # By Default
    search_kwargs = {
        "k": 2,
    }

)
```

```python
# Step 5: User Query on retriever based on our document and get result
query = "What is LangChain?"

# get the result
results = retriever.invoke(query)
```

```python
# print the retriever result
for i , doc in enumerate(results, start=1):
    print(f"Result: {i}:\n {doc.page_content}")
```

### Find the same result using similarity search

- this is a one way to get the result but we can't integrate into Chain by using similarity search

```python
results = vectorstore.similarity_search(query, k=2)
for i , doc in enumerate(results, start=1):
    print(f"Result: {i}:\n {doc.page_content}")
```

###  4. Complete Summary of Similarity Search retriever

```python
"""  
┌──────────────────────────────────────────────────────────────────┐
│                SIMILARITY SEARCH RETRIEVER                       │
│                                                                  │
│  WHAT:  Finds top-k documents whose embeddings are closest       │
│         to the query embedding (meaning-based matching)          │
│                                                                  │
│  HOW:   query → embed → compare with all doc vectors →           │
│         rank by distance → return top-k                          │
│                                                                  │
│  METRICS:                                                        │
│    Cosine Similarity → Angle between vectors (default-ish)       │
│    Dot Product        → Angle X magnitude                        │
│    Euclidean (L2)     → Straight-line distance (FAISS default)   │
│                                                                  │
│  SYNTAX (Retriever):                                             │
│    retriever = vectorstore.as_retriever(                         │
│        search_type="similarity",                                 │
│        search_kwargs={"k": 4, "filter": {...}}                   │
│    )                                                             │
│    docs = retriever.invoke("query")                              │
│                                                                  │
│  SYNTAX (Direct):                                                │
│    vectorstore.similarity_search(query, k=4)                     │
│    vectorstore.similarity_search_with_score(query, k=4)          │
│                                                                  │
│  KEY PARAMS:                                                     │
│    k       → Number of results (default: 4)                      │
│    filter  → Metadata pre-filter (search only matching subset)   │
│                                                                  │
│  FAISS SPECIFICS:                                                │
│    save_local("path")     → Persist index                        │
│    load_local("path",     → Reload without re-embedding          │
│      embeddings,                                                 │
│      allow_dangerous_deserialization=True)                       │
│    Returns L2 DISTANCES → lower score = MORE similar             │
│                                                                  │
│  STRENGTHS:                                                      │
│    ✅ Fast (ms even at millions of docs)                          │
│    ✅ Simple to configure                                         │
│    ✅ Semantic understanding (no keyword dependence)              │
│    ✅ Default search_type in LangChain                            │
│                                                                    │
│  WEAKNESSES:                                                       │
│    ❌ Can return redundant/near-duplicate chunks                  │
│    ❌ No score cutoff (may include weak matches)                  │
│    ❌ Fails on exact IDs/codes (use BM25 for that)                │
│    ❌ Quality capped by embedding model + chunking                │
│  WHEN TO USE:                                                     │
│    ✔ Basic RAG / Q&A systems                                      │
│    ✔ When speed matters                                           │
│    ✘ Avoid when you need diverse results → use MMR                │
│    ✘ Avoid for keyword-heavy search → use Hybrid                  │
│                                                                   │
│  WHEN TO USE vs ALTERNATIVES:                                     │
│    similarity                  → Default choice, simple RAG       │
│    mmr                         → Need diverse results             │
│    similarity_score_threshold  → Need quality control             │
│    Ensemble (BM25 + Vector)    → Keyword-heavy domains            │
│                                                                  │
│  PIPELINE:                                                       │
│    [Loader] → [Splitter] → [Embeddings] → [FAISS]                │
│                                                   ↓              │
│    Query → [Embed] → [SIMILARITY SEARCH] → Top-k → [LLM]         │
│                          THIS COMPONENT                          │
│                                                                  │
│  GOLDEN RULE:                                                    │
│  "Start with similarity, k=4. If results feel redundant,         │
│   switch to mmr. If results feel noisy, add a score threshold    │
│   or a re-ranker."                                               │
└──────────────────────────────────────────────────────────────────┘

"""
```
