## Topic: Introduction of LangChain Models Component

### Introduction of LangChain Models Component

- Definition:
    - The Model Component in LangChain is a crucial part of the framework, designed to facilitate interactions with various Language models and embedding models.



- Key Note:
    - LangChain Models Component Provide a common interface, so that we interaction with various type of AI models (language models, embedding models)



```python
"""  
Without LangChain:
  OpenAI SDK  → openai.ChatCompletion.create(...)
  Anthropic   → anthropic.messages.create(...)
  Google      → genai.GenerativeModel(...).generate(...)
    
    - All different APIs!

With LangChain:
  llm.invoke("Hello")  ← Same method, ANY provider 


"""
```

### Types of Models in LangChain
- Model Types
    - 1. Language Model
        - LLMs
        - Chat Model

    - 2. Embedding Model


---------------------------------------------------------------------------

- 1. Language Model
    - 1.1: LLMs (Large Language Models)
        - A language model that processes text and generates text.

        - input: Text (string)
        - output: Text(string)

        - Use case:
            - Simple text completion

    
    - 1.2: Chat Model
        - A Chat Model is designed to work with messages rather than treating everything as one plain text string.

        - Designed for conversational interactions using messages such as system, human, and AI.

        - Example: 
            - GPT, Gemini, Claude.


        - use:
            - Use for creating chatbot 


## 2. Embedding Model
- Models that convert text into numerical vectors (arrays of floats). These vectors capture the semantic meaning of text, enabling similarity search.


- input: 
    - Text

- output:
    - Vector


- Use case:
    - Mainly used for semantic search and RAG base application.

```python
# Example of Embedding model
"""   
Text Input                    Vector Output (1536 dimensions)
─────────────                 ────────────────────────────────
"The cat sat on the mat"  →  [0.012, -0.034, 0.078, ..., 0.045]
"A kitten rested on rug"  →  [0.011, -0.033, 0.079, ..., 0.044]  ← Very similar!
"Stock market crashed"    →  [-0.089, 0.045, -0.002, ..., 0.091]  ← Very different!


"""
```

- Summary of Models Component in LangChain:

    - It abstracts the complexity of working directly with different LLMs, chat models, and embedding models, providing a uniform interface to communicate with them. This makes it easier to build applications that rely on AI-generated text, embeddings for similarity search, and retrieval-augmented generation(RAG).
   
    - Every LangChain application — whether it's a chatbot, RAG pipeline, or autonomous agent — ultimately sends data to a Model and gets a response back. The Model is the engine; everything else is plumbing.
