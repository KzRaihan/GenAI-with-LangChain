### Topic: Language Model

### Introduction of Language Models

- Definition:
    - Language models are AI systems designed to process, generate, and understand natural language text.


- Type of Language Models
    - 1. LLMs

    - 2. Chat Models


### 1. LLMs (Large Language Models)
- Definition:
    - A language model that processes text and generates text.


- Characteristic: 
    - General purpose models that is used for raw text generation.
    - They take a string(or plain text) as input and returns a string(plain text).
    - These are traditionally older models and are not used much now.


- Application:
    - text generation
    - text summarization
    - code generation
    - question answering

### 2. Chat Models
- Definition:
    - 


- Characteristic: 
    - Language models that are specialized for conversational tasks.
    
    - They take a sequence of messages as inputs as return chat messages as outputs(as opposed to using plain text)

    - These are traditionally newer models and used more in comparison to the LLMs.


- Application:
    - Chatbot
    - coding assistant
    - agent
