- What does invoke() mean?
    - In LangChain, invoke() means: "send this input to the model and get the result back."

- In simple words:
    - invoke() = call/run the model with an input.

```python
# library for OpenAI chat models
from langchain_openai import ChatOpenAI


from dotenv import load_dotenv

# Load environment variables from .env
load_dotenv()

# Create an ChatOpenAI instance with some parameters

model = ChatOpenAI(
    model="gpt-4",               # Specify the model name
    temperature=0.7,             # Set the temperature for randomness (creative responses)
    max_completion_tokens=100,   # Limit the maximum number of tokens in the response
)


# Invoke the model with a prompt
response = model.invoke("What is the capital of Bangladesh?")

print("Response from the model:", response.content)
```

- 1. model="gpt-4",
    - This is the model that we use to Perform doing this work

- 2. temperature = 0.7
    - temperature is a parameter that controls the randomness of a language models's output.

    - It affects how creative or deterministic the response are.

    - lower value
        - (0.0 - 0.3) -> More deterministic and Predictable.

    - Higher value
        - (0.7 - 1.5) -> More random Creative and diverse response.

    - Use case:
        - Factual answer(math, code ,facts) -> (0.0 - 0.3)
        - Balanced response(general QA, explanations) -> (0.5-0.7)
        - Creative writing storytelling, jokes -> (0.9-1.2)
        - Maximum randomness (wild ideas, brainstorming) - (1.5+)
