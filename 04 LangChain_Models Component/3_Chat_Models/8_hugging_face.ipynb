## Topic: Open Source Models

- Definition:
    - Open-source language models are freely available AI models that can be downloaded, modified, fine-tuned, and deployed without restrictions from a central provider.
    
    - Unlike closed-source models such as OpenAI's GPT-4, Anthropic's Claude, or Google's Gemini, open-source models allow full control and customization.


### Open-Source Models
## ========================================

- 1. Cost
    - Free to use (no API Cost).

- 2. Control:
    - Can modify, fine tune, and deploy anywhere.

- 3. Data Privacy:
    - Runs locally (no data sent to external servers)


- 4. Customization:
    - Can fine-tune on specific datasets.

- 5. Deployment:
    - Can be deployed on on-premise servers or cloud.



### Close-Source Models
## ========================================

- 1. Cost
    - Paid API usage(eg. OpenAI changes per token).

- 2. Control:
    - Locked to provider's infrastructure.

- 3. Data Privacy:
    - Sends Queries to provider's servers.


- 4. Customization:
    - No access to fine-tuning in most cases.

- 5. Deployment:
    - Must use vendor's API.

### Ways to use open-source Models

- Two ways
    - 1. Using HF Inference API
        - Create an Hugging face API key () then Create new tokens
        - Step1: Go to hugging face and crate an account
        - step2: Go to AccessTokens
        - step3: Create new Access Token
        - step5: Read mode
        - step6: Tokens name (LangChain_HF_Token)
        - step7: Create token
        - step8: copy your token and done
        

    - 2. Running Locally
        - step1: Download the model
        - step2: use the model


### Disadvantage of Open-source models

- 1. High Hardware Requirements:
    - Running large Models (eg. LLaMa-2.70B) requires expensive GPUs.

- 2. Setup Complexity:
    - Requires installation of dependencies like PyTorch, CUDA, transformers.

- 3. Lack of RLHF:
    - Most open-source models don't have fine-tuning with human feedback making them weaker in instruction-following.

- 4. Limited Multi-modal Abilities:
    - Some of Open Models do't support images, audio, or video like GPT-4V.


```python
from langchain_huggingface import ChatHuggingFace, HuggingFaceEndpoint


# Load variables from the .env file
load_dotenv()

# Create the Hugging Face endpoint
endpoint = HuggingFaceEndpoint(
    repo_id="mistralai/Mistral-7B-Instruct-v0.2", # Specify the model repository ID
    task="text-generation", # Specify the task type
    temperature=0.7,        #  
    max_new_tokens=512, # 
    do_sample=True,
)

# Create the Hugging Face chat model
model = ChatHuggingFace(llm=endpoint)
```
