### 1. What is GenAI?
- Definition:
    - Generative Ai is a type of artificial intelligence that creates new content such as text, images, music, or code by learning patterns form existing data, mimicking human creativity.


- What is the different from the ML?

- For Machine Learning(ML):
    - We create a model for classification and regression type of tasks.
    
    - ML model can only work with specific type of problem like 
    - some of ML model:
        - stock Price Prediction model(Regressor type ml model)
        - cat and dog classification model (classification type ml model)
        - Ranking type problem__Similar  product ranking (Recommendation type of ml model)

    - Dropback of ML model:
        - Traditional machine learning models cannot create new content because they are designed only to recognize patterns and make predictions from existing data.

- For Generative AI (GenAI):
    - GenAI creates new content such as text, images, music, or code by learning patterns form existing data
    
    - and also try to mimicking human creativity.
