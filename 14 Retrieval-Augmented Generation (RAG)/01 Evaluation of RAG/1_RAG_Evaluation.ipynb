### 3. Supervised Fine-Tuning (SFT) for LLMs

- Definition: 
    - Supervised Fine-Tuning (SFT) is a process of taking a pre-trained language model and further training them on a smaller, task-specific dataset with labeled examples. 
    
    - Its goal is to adjust weights of pre-trained model so that it performs better on our specific task without losing its general knowledge acquired during pre-training.

- link:
    - https://www.geeksforgeeks.org/artificial-intelligence/supervised-fine-tuning-sft-for-llms/



```python
""" 
- How Does Supervised Fine-Tuning Work?

    - Step1: Collect Data (Private / Personal)
        - A few hundred - few hundred-thousand carefully curated example (prompt -> desired output)
    
    - Step2: Choose Method
        - Full Fine-Tuning,  Partial Fine-Tuning, Low-Rank Adaptation (LoRA), QLoRA, Parameter-Efficient Fine-Tuning (PEFT).

    - Step3: Train for a few epochs
        - A lower epochs is used to make gradual adjustments to the model's weights.
        - This helps preserve previously learned knowledge while allowing the model to adapt to the new task.

    
    - Step4: Evaluate and safety-test
        - The model is tested on the target task to measure its performance.
        
        - Based on the results, additional layers can be fine-tuned or training parameters can be adjusted to improve accuracy.

        - see the Measure exact-match, factuality, and hallucination rate against held-out data.

        
- Key Rule of thumb:
    - By applying the Supervised Fine-Tuning, we solve the 
        
        - i.   Response of Private/Person Data.
        - ii.  Response of Recent Data (here we only solve domain-specific data recent data relative query).
        - iii. overcome of the Hallucination of LLM.

    - There are so major problems of Fine-tuning  
        - i.   High Computational Cost
        - ii.  Need to Strong Technique Expatriates to doing fine-tuning.
        - iii. Can't give the best response of Frequently changing data. 

    - Solution:
        - In-Context-Learning Technique

"""
```

### 4. In-Context-Learning Technique

- Definition:
    - In-Context Learning (ICL) is a technique where we teach an LLM how to perform a task by giving examples inside the prompt, without changing the model's parameters or training it.


    - In-Context Learning is a core capability of Large Language Models (LLMs) like GPT-3/4, Claude, and Llama, where the model learns to solve a task purely by seeing examples in the prompt—without updating its weights.


    - Instead of fine-tuning the model, we show the model examples in the prompt and ask it to follow the demonstrated pattern.

- Key note:
    - In-Context-Learning is an "emergent property" of an LLM.

    - emergent property:
        - An emergent property is a behaviour or ability that suddenly appears in a system when it reaches a certain scale or complexity—even though it was not explicitly programmed or expected from the individual components.

#### Key Rule of Thump of In-Context-Learning Technique
- To get the best result of the LLm without fine-tuning, we use prompting technique(Few-short prompting).

- Few-short prompting means
    - an AI technique where we give a model a small number of examples (usually two to five) inside our prompt to show it the exact pattern, style, or format we want.

    -  By learning from these examples, the model can better understand the task and generate more accurate and relevant outputs without additional training.
