## Topic: Practical Implementation of RecursiveCharacterTextSplitter

### Example 1: Basic Example

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter


# =========================================================
# STEP 1: Create sample text
# =========================================================

text = """
Machine learning is a branch of artificial intelligence.
It allows computers to learn patterns from data.

Deep learning is a subset of machine learning.
It uses artificial neural networks with multiple layers.

Convolutional Neural Networks are commonly used
for image classification and computer vision tasks.
"""


# =========================================================
# STEP 2: Create the Recursive Character Text Splitter
# =========================================================

text_splitter = RecursiveCharacterTextSplitter(
    chunk_size=200,
    chunk_overlap=50
)


# =========================================================
# STEP 3: Split the text
# =========================================================

chunks = text_splitter.split_text(text)


# =========================================================
# STEP 4: Display the chunks
# =========================================================

print(f"Number of chunks: {len(chunks)}")

for i, chunk in enumerate(chunks):

    print("\n" + "=" * 60)
    print(f"CHUNK {i + 1}")
    print("=" * 60)

    print(chunk)
```

### Example 2: With Custom Parameters

```python
# Example 2: With Custom Parameters
from langchain_text_splitters import RecursiveCharacterTextSplitter

text = """
LangChain is a framework for developing applications powered by large 
language models (LLMs). It provides a standard interface for working 
with different LLM providers, making it easy to switch between models.

The framework includes several key components: prompts, chains, agents, 
memory, and retrievers. Each component can be used independently or 
combined to build complex AI applications.

One of the most popular use cases for LangChain is Retrieval-Augmented 
Generation (RAG). RAG combines the knowledge of an LLM with external 
data sources to provide accurate, up-to-date answers to user questions.

LangChain also supports tool use, allowing LLMs to interact with 
external APIs, databases, and other systems. This enables the creation 
of autonomous AI agents that can perform complex tasks.
"""

splitter = RecursiveCharacterTextSplitter(
    chunk_size=100,
    chunk_overlap=20,
    # separators=["\n\n", "\n", ". ", "! ", "? ", "; ", ", ", " ", ""], # ←  Default
    separators=["\n\n", "\n", ". ", " ", ""],
    length_function=len,
    keep_separator=True,
    add_start_index=True,      # ← Track position in original text
    strip_whitespace=True
)

chunks = splitter.split_text(text)
print(f"Length of chunks : {len(chunks)}")

# Each chunk now
for chunk in chunks:
    # print(f"Start: {chunk.metadata.get('start_index', 'N/A')}")
    print(f"Length: {len(chunk)}")
```

### Real-World Example  — PDF RAG Pipeline 

```python
# Example : RAG PIPELINE without DATABASE
# workflow:  Local Data Source(PDF) -> Chunk -> LLM(summarize) -> output parser

import os
from dotenv import load_dotenv

# Import necessary LangChain modules
from langchain_community.document_loaders import PyPDFLoader
from langchain_text_splitters import RecursiveCharacterTextSplitter
from langchain_openai import ChatOpenAI
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser

# Load environment variables (OPENAI_API_KEY)
load_dotenv()

# =========================================================
# STEP 1: Load the PDF file
# =========================================================
pdf_path = "../data/dl-curriculum.pdf"

# Verify file exists before loading
if not os.path.exists(pdf_path):
    raise FileNotFoundError(f"PDF file not found at: {pdf_path}")

loader = PyPDFLoader(pdf_path)
docs = loader.load()
print(f"📄 Loaded {len(docs)} pages from PDF.")
```

```python
# =========================================================
# STEP 2: Split Document into Chunks
# =========================================================
splitter = RecursiveCharacterTextSplitter(
    separators=["\n\n", "\n", ". ", " ", ""],
    chunk_size=1000,      # 300 is too small for summaries; 1000 provides better context
    chunk_overlap=100,
)

chunks = splitter.split_documents(docs)
print(f"✂️ Total Chunks Created: {len(chunks)}")

# Helper: Combine all chunk contents into a single string for the prompt
full_text_content = "\n\n".join(chunk.page_content for chunk in chunks)
```

```python
# =========================================================
# STEP 3: Create Prompt, Model, and Parser
# =========================================================
prompt = ChatPromptTemplate.from_messages([
    ("system", "You are an expert academic assistant. Provide a concise, well-structured summary of the curriculum."),
    ("human", "Summarize the following curriculum content:\n\n{docs}")
])

# Specify model and low temperature for accurate summarization
model = ChatOpenAI(model="gpt-4o-mini", temperature=0.2)

# Output parser converts AIMessage directly to string
parser = StrOutputParser()
```

```python
# =========================================================
# STEP 4: Create & Run the Chain
# =========================================================
chain = prompt | model | parser

# Invoke the chain with the combined text
response = chain.invoke({"docs": full_text_content})

print("\n" + "=" * 60)
print("📝 SUMMARY OF LOADED DOCUMENT:")
print("=" * 60)
print(response)
```
