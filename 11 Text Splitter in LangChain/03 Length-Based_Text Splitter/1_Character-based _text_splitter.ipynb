### 1. Introduction of Length Based Text Splitter

- Definition:
    - A Character-based Text Splitter divides text into smaller chunks according to a specified number of characters.

### 2. Why Do We Need Character-based Splitting?

- Problem 1 — Documents can be very large.

- Problem 2 — Better retrieval granularity.

- Problem 3 — LLM context limitations.



### 3. Pros and Cons of Character-based Text Splitter

- 1. Pros:
- ===========================
    - 1. Simple
        - It is relatively easy to understand and configure.
    
    - 2. Fast
        - Character-oriented splitting is computationally lightweight compared with more sophisticated semantic approaches.
    
    - 3. Useful for simple text
        - Plain text files
        - Simple documents
        - Logs
        - Basic notes
        - Simple knowledge bases

- 2. Cons:
- ===========================
    - 1. It doesn't understand semantics

    - 2. Important sentences can be separated

    - 3. Not ideal for every document structure

    - 4. Chunk size needs experimentation


### 4. Basic Practical Example

```python
from langchain_text_splitters import CharacterTextSplitter

# Text 
text = """
Space exploration has led to incredible scientific discoveries.
From landing on the Moon to exploring Mars, humanity
continues to push the boundaries of what's possible beyond our
planet.

These missions have not only expanded our knowledge of the
universe but have also contributed to advancements in
technology here on Earth. Satellite communications, GPS, and
even certain medical imaging techniques trace their roots back
to innovations driven by space programs.

"""


# create an CharacterTextSplitter object:
splitter = CharacterTextSplitter(
    separator = "",
    chunk_size = 100,
    chunk_overlap = 0,

)

# split the text using split_text() function
chunk = splitter.split_text(text)

print(f"Total Chunk : {len(chunk)}")
print(f"Type of Chunk: {type(chunk)}")
```

```python
# print one chunk data or text
print(f"1st Chunk Data: \n{chunk[0]}")
```

### 5. Practical Example — Connect document loader With Character splitter in LangChain

```python
from langchain_community.document_loaders import PyPDFLoader
from langchain_text_splitters import CharacterTextSplitter

# Load one PDF file
loader = PyPDFLoader("../data/dl-curriculum.pdf")

# Load all pages
docs = loader.load()

# Print total pages
print("Total pages:", len(docs))

# Print the 2nd page
print(docs[1].page_content)
```

```python
## Now Apply the splitter

# Configure the CharacterTextSplitter by define an object
splitter = CharacterTextSplitter(
    separator = "",
    chunk_size = 100,
    chunk_overlap = 0,

)

# chunk of the document(loader)
chunk = splitter.split_documents(docs)

print(f"Total Chunk: {len(chunk)}")

print(f"second chunk Data: \n{chunk[2].page_content}")
```

```python
""" - What Happened Internally?

                 Large Text
                     │
                     ▼
             Find separators
                     │
                     ▼
             Create smaller
                text pieces
                     │
                     ▼
            Respect chunk size
                     │
                     ▼
            Apply overlap
                     │
                     ▼
                  Chunks

 """
```

```python
# =========================================================
# Example: TextLoader + CharacterTextSplitter
# =========================================================


from langchain_community.document_loaders import TextLoader
from langchain_text_splitters import CharacterTextSplitter


# =========================================================
# STEP 1: Load the text file
# =========================================================

loader = TextLoader(
    "../data/good.txt",
    encoding="utf-8"
)

documents = loader.load()


# =========================================================
# STEP 2: Create Character-based Text Splitter
# =========================================================

text_splitter = CharacterTextSplitter(
    separator="\n\n",
    chunk_size=100,
    chunk_overlap=20
)


# =========================================================
# STEP 3: Split Documents into smaller chunks
# =========================================================

chunks = text_splitter.split_documents(documents)


# =========================================================
# STEP 4: Display statistics
# =========================================================

print(f"Original Documents: {len(documents)}")
print(f"Generated Chunks: {len(chunks)}")


# =========================================================
# STEP 5: Inspect each chunk
# =========================================================

for i, chunk in enumerate(chunks):

    print("\n" + "=" * 60)
    print(f"CHUNK {i + 1}")
    print("=" * 60)

    print("Content:")
    print(chunk.page_content)

    print("\nMetadata:")
    print(chunk.metadata)
```

### Complete Summary

- What is Character-based Text Splitter?
    - A text splitter that divides text into smaller chunks using character-oriented length and configured separators.

===================================================================================================

- Why do we need it?
    - Because large documents are difficult and inefficient to process as single units for:
        - Embeddings
        - Retrieval
        - RAG
        - LLM context

===================================================================================================

- Main parameters
    - 1. separator
        - Determines where the text can be split
    
    - 2. chunk_size
	    - Target/max size according to the configured length function
    
    - 3. chunk_overlap	
        - Shared content between adjacent chunks

===================================================================================================

- Main methods:
    - 1. split_text(text)
        - → returns List[str]

    - 2. split_documents(documents)
        - → returns List[Document]

===================================================================================================
#### Key takeaway: 
- CharacterTextSplitter is a simple and useful starting point for understanding chunking, but for general-purpose RAG, you should next study RecursiveCharacterTextSplitter and understand exactly how its separator hierarchy improves upon basic character-based splitting.
