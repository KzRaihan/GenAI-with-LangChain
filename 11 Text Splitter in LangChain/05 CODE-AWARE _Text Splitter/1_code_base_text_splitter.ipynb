## Topic: Code Base Text Splitter

### 1. Introduction of Code Base Text Splitter

- Definition:
    - Code Splitters split source code files while respecting the programming language's syntax. Instead of cutting in the middle of a function, class, or method, they split at logical code boundaries.


- Why Use Code-Aware Splitting?
    - If we use RecursiveCharacterTextSplitter on code, it can cut a function in half. That creates useless chunks for RAG.



- When to Use Code Splitters?
    - Building a Codebase RAG (chat with your codebase)
    - Creating embeddings for source code search
    - Building AI coding assistants that reference your code
    - Indexing GitHub repositories

```python
## Example
from langchain.text_splitter import RecursiveCharacterTextSplitter,Language

text = """
class Student:
    def __init__(self, name, age, grade):
        self.name = name
        self.age = age
        self.grade = grade  # Grade is a float (like 8.5 or 9.2)

    def get_details(self):
        return self.name"

    def is_passing(self):
        return self.grade >= 6.0


# Example usage
student1 = Student("Aarav", 20, 8.2)
print(student1.get_details())

if student1.is_passing():
    print("The student is passing.")
else:
    print("The student is not passing.")

"""

# Initialize the splitter
splitter = RecursiveCharacterTextSplitter.from_language(
    language=Language.PYTHON,
    chunk_size=300,
    chunk_overlap=0,
)

# Perform the split
chunks = splitter.split_text(text)

print(len(chunks))
print(chunks[1])
```

### Example 1: Splitting Python Code

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter, Language

python_code = """
import pandas as pd
import numpy as np

class DataProcessor:
    def __init__(self, data):
        self.data = data
    
    def clean(self):
        # Remove null values and duplicates
        self.data = self.data.dropna()
        self.data = self.data.drop_duplicates()
        return self.data
    
    def normalize(self):
        min_val = min(self.data)
        max_val = max(self.data)
        normalized = [(x - min_val) / (max_val - min_val) for x in self.data]
        return normalized

def load_csv(filepath):
    df = pd.read_csv(filepath)
    return df.to_dict('records')

def calculate_average(numbers):
    if not numbers:
        return 0
    return sum(numbers) / len(numbers)

if __name__ == "__main__":
    print("Data Processing Module")
"""

# Create Python-aware code splitter
python_splitter = RecursiveCharacterTextSplitter.from_language(
    language=Language.PYTHON,
    chunk_size=500,      # Keep functions/classes reasonably sized
    chunk_overlap=40
)

chunks = python_splitter.split_text(python_code)

print(f"Total Chunk: {len(chunks)}\n\n")

for i, chunk in enumerate(chunks):
    print("=" * 60)
    print(f"CHUNK {i+1} | {len(chunk)} CHARACTERS")
    print("=" * 60)
    print(chunk)
    print()
```

### Example 2: Splitting JavaScript Code

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter, Language

js_code = """
function calculateSum(a, b) {
    return a + b;
}

class User {
    constructor(name, email) {
        this.name = name;
        this.email = email;
    }
    
    getUserInfo() {
        return `Name: ${this.name}, Email: ${this.email}`;
    }
}

const greetUser = (userName) => {
    console.log(`Hello, ${userName}! Welcome back.`);
};

export { calculateSum, User, greetUser };
"""

js_splitter = RecursiveCharacterTextSplitter.from_language(
    language=Language.JS,
    chunk_size=400,
    chunk_overlap=30
)

chunks = js_splitter.split_text(js_code)
print(f"Total JS chunks: {len(chunks)}")
```
